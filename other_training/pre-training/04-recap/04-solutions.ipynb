```python
# read in the csv file with filepath defined below to a pandas dataframe
# store your results in a variable called df
# hint: remember to use pd. before the function you are using to read in the csv file

filepath = "./data/st_indicators_data.csv"

df = pd.read_csv(filepath)
```

```python
# inspect your data so that you know its shape, size, what its indexed on, and what the column names are
# solution below

df.head()
```

```python
# use the pandas melt method to reshape your data into long format
# solution below

melt_df = df.melt(id_vars="indicator_variable", var_name="year", value_name="market_share")
melt_df.head()
```

```python
# change the index to be your year column
# solution below

melt_df = melt_df.set_index("year")
melt_df.head()
```

```python
# use the pandas pivot method to get the indicator_variable values along the top as columns
# solution below

pivot_df = melt_df.pivot(columns="indicator_variable", values="market_share")
pivot_df.tail()
```

```python
# define a new column which contains the difference in market share between our CB7 pathway and the ZEV mandate
# solution below

pivot_df["cb7_zm_difference"] = pivot_df["cb7_deployment"] - pivot_df["zev_mandate"]
```

```python
# find the biggest gap between the two i.e. the maximum value in your new column
# solution below

max_difference = pivot_df["cb7_zm_difference"].max()

print(f"The biggest gap is {100 * max_difference:.1f} pp.")
```

```python
# find the first year at which the ZEV mandate and CB7 pathway market shares are equal
# try to do this programmatically (by writing code), not just by looking at the data!
# solution below

equal_condition = pivot_df["zev_mandate"] == pivot_df["cb7_deployment"]
first_year_equal = pivot_df[equal_condition].index[0]

print(f"The first year in which the CB7 and ZEV mandate values match is {first_year_equal}.")
```

```python
# define a function which takes a dataframe as its input and returns a dataframe after the application
# of one of the operations covered in this recap

# example solution below

def create_difference_column(df : pd.DataFrame,
                             first_column : str,
                             second_column : str
                             ) -> pd.DataFrame:
    
    """
    Function to create a new column in dataframe
    with the difference between two input columns.

    Args:
        df (pd.DataFrame): input dataframe
        first_column (str): name of first column
        second_column (str): name of second column

    Returns:
        pd.DataFrame with new column added on
    """

    df[f"diff_{first_column}_{second_column}"] = df[first_column] - df[second_column]

    return df

# testing out function

create_difference_column(pivot_df, "cb7_deployment", "zev_mandate")
pivot_df.tail()
```
