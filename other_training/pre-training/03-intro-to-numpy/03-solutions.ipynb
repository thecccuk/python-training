Question 1a) Define three parameters for creating a range of years:
- ```START_YEAR```: for the start of the period
- ```END_YEAR```: for the end of the period
- ```INTERVAL``` for the timegap between sequential years in the range

````python
# example solution below

START_YEAR = 2030
END_YEAR = 2060
INTERVAL = 5
````

Question 1b) Generate a numpy array which contains the years defined by these three parameters
- Call the array variable ```years```
- Use the ```np.arange()``` method. You can look up the syntax for this online.
- Run ```print(years)``` to make sure this gives you what you expect. 

````python
# solution below

years = np.arange(start=START_YEAR,
                  stop=END_YEAR+INTERVAL,
                  step=INTERVAL)

print(years)
````

Question 1c) Generate another array of same length with some numerical data.
- Use ```len(years)``` to guarantee the array has same number of values
- Use ```np.random.randint()```
- Make the numbers generated be between 3e8 and 4e8
- Call this array ```emissions```

````python
# solution below

baseline_emissions = np.random.randint(3e8, 4e8, len(years))
````

Question 2a) Store the value for the first element in the baseline_emissions array.
- Hint: remember indexing!
- Store this in a variable called ```starting_emissions```.

````python
# solution below

starting_emissions = baseline_emissions[0]
````

Question 2b) Find the mean of the ```baseline_emissions``` values. 
- Store this in a variable called ```average_emissions```.

````python
# solution below

average_emissions = baseline_emissions.mean()
````

Question 2c) Work out the difference between your ```average_emissions``` and ```starting_emissions``` variables.
- Store this as ```difference_to_average```.

````python
# solution below

difference_to_average = starting_emissions - average_emissions
````

Question 2d) Generate a new array with only the values of ```emissions``` which are higher than ```average_emissions```.
- Call this array ```high_emissions```
- Hint: remember masking and indexing!

````python
# solution below

# explicitly defining the mask first
mask = baseline_emissions > average_emissions
high_emissions = baseline_emissions[mask]

# or doing all of this in one step
high_emissions = baseline_emissions[baseline_emissions > average_emissions]
````

Question 3a) Convert the values in your ```baseline_emissions``` array and ```starting_emissions``` variable from MtCO2e -> kgCO2e.
- First define a numerical value for your ```conversion_factor```
- Then divide your ```emissions``` array by this value, making sure to store the result in the variable.

````python
# solution below

conversion_factor = 1e6

baseline_emissions = baseline_emissions / conversion_factor
starting_emissions = starting_emissions / conversion_factor
````

Question 3b) Generate a new array which interpolates linearly between ```starting_emissions``` and **zero** across your defined time period.
- Call this array ```simple_nz_pathway```
- Hint: use ```np.linspace()```

````python
# solution below

simple_nz_pathway = np.linspace(start=starting_emissions,
                                stop=0,
                                num=len(years))

simple_nz_pathway
````

Question 3c) Generate a new array which interpolates linearly between ```starting_emissions``` and **-100** across your defined time period.
- Call this array ```simple_net_negative_pathway```
- Hint: use ```np.linspace()``` again, but change the ```stop``` argument.

````python
# solution below

simple_net_negative_pathway = np.linspace(start=starting_emissions,
                                stop=-100,
                                num=len(years))

simple_net_negative_pathway
````

Question 4a) We have four equal length numpy arrays. Let's make them into a 2D array.
- Hint: use ```np.column_stack()```
- Call this ```emissions_data```

````python
# solution below

emissions_data = np.column_stack([years,
                                  baseline_emissions,
                                  simple_nz_pathway,
                                  simple_net_negative_pathway])

emissions_data
````

Question 4b) Pass this array into a pandas dataframe.
- Use ```pd.Dataframe()```
    * Make sure you specify both the ```data``` and ```columns``` arguments.
    * Use these column names: ["years", "baseline_emissions", "net_zero_pathway", "net_negative_pathway"]
- Store this as ```emissions_df```.

````python
# solution below

# define column names in a list for ease first
column_names = ["years",
                "baseline_emissions",
                "net_zero_pathway",
                "net_negative_pathway"]

emissions_df = pd.DataFrame(data=emissions_data, columns=column_names)
````

Question 4c) Set the ```years``` column as your index for ease.
- Hint: use the ```pd.DataFrame.set_index()``` method.
- Make sure you assign the result back to your ```emissions_df``` variable.

````python
# solution below

emissions_df = emissions_df.set_index("years")
````
