# 3 Dictionaries and Pandas

## 3.1 Dictionaries

Dictionaries are mutable objects used for storing _key-value pairs_.

Dictionaries are constructed with curly braces: `{}`.

Each _key_ is mapped to a corresponding _value_ with a colon `:`.

Key-value pairs are separated by commas:

```
# Create a sector with name, emissions and status
sector = {
    "name": "Surface transport",
    "emissions": 99,
    "largest emitter": True
}
# It is not necessary to put each key-value pair on a separate line
# but it can help with readability
type(sector)
```

This dictionary has three _key-value_ pairs.

The keys are `"name"`, `"emissions"`, and `"largest emitter"`.

The values corresponding to these keys are `"Surface transport"`, `99`, and `True`.

Our dictionary keys _don't have to be_ strings, though they almost always are.
Our dictionary values can be anything we like.

### 3.1.1 Extracting values

_Like_ lists and arrays, we extract values from dictionaries with square
brackets `[]`.

_Unlike_ lists and arrays we extract values from dictionaries with _keys_ not
_indices_.

For example, if you wanted to know the sector's emissions you could use the
`"emissions"` key as:

```
sector["emissions"]
```

Similarly, you could check whether it is the largest emitting sector with:

```
sector["largest emitter"]
```

## 3.2 Pandas

The Pandas library is a data-manipulation package built on top of NumPy.

Its aim is to provide high performance, easy-to-use data structures and data
analysis tools.

By _convention_ we _alias_ the pandas package to `pd` at import:

```
import pandas as pd
```

Pandas has two key data structures: the `Series` and the `DataFrame`.

* `Series` objects are like a hybrid of a dictionary and a one-dimensional
  NumPy array.

* `DataFrame` objects are like a hybrid of a dictionary and a _two_-dimensional
  Numpy array.

## 3.3 Pandas Series

The pandas `Series` object is a _labelled_ one-dimensional array. In general,
we create a `Series` as: `series = pd.Series(data, index=index)`

`data` may be a list, dictionary, or one-dimensional NumPy array.

The `index` is _optional_, and represents labels which point to elements
of the `Series`.

If no `index` is specified then Pandas will choose some sensible labels:

```
# Create a Series of the numbers 0.1, 0.2 and 0.3
series = pd.Series([0.1, 0.2, 0.3])
series
```

Alternatively, we can specify an `index` to use:

```
# Include a user-specified index
series = pd.Series([0.1, 0.2, 0.3], index=["a", "b", "c"])
series
```

The `pd.Series()` function can be used to create `Series` objects from
dictionaries:

```
# Create a Series from a dictionary
exciting_dict = {"d": 0.1,
                 "e": 0.2,
                 "f": 0.3}
exciting_series = pd.Series(exciting_dict)
exciting_series
```

## 3.4 Pandas `DataFrame`s

The pandas `DataFrame` object is reminiscent of the all-too-familiar
spreadsheet.

Here rows typically represent observations, and columns typically represent
variables. For example:

| nation   | population | emissions |
| -------- | ---------- | --------- |
| England  | 56490000   | 330.5     |
| Scotland | 5480000    | 39.8      |
| Wales    | 3110000    | 34.2      |
| Ireland  | 5120000    | 54.9      |

We're going to construct a pandas `DataFrame` to store this data on UK nations.

Let's start by creating a dictionary called `nations` containing three key-value
pairs (representing our three variables):

```
# Dictionary with four nations
nations = {
    "nation": ["England", "Scotland", "Wales", "Ireland"],
    "population": [56490000, 5480000, 3110000, 5120000],
    "emissions": [330.5, 39.8, 34.2, 54.9]
}
```

We can then convert this dictionary to a `DataFrame` with `pd.DataFrame()`:

```
# Convert our dictionary to a DataFrame
df = pd.DataFrame(nations)
df
```

See that the keys of our dictionary have become columns in our `DataFrame`.

The `.set_index()` method allows us to change the index of our `DataFrame`.

We may pass a list or array to `.set_index()`, or the name of an existing
column:

```
# Change the row index to nation name
df = df.set_index("nation")
df
```

We can use the `.shape` attribute to check the number of rows and columns
in our `DataFrame`:

```
df.shape
```

### 3.4.1 Extracting data

The Python and NumPy index operators `[]`, can be used to extract values from
`Series` and `DataFrame` objects:


```
# Extract the emissions column
df["emissions"]
```

Pandas also provides the `.loc` and `.iloc` attributes for optimised
data access:

* `.loc` is primarily used for _label_-based extraction.

* `.iloc` is primarily used for _index_-based extraction.

We prefer `.loc` and `.iloc` in all but simple column extraction operations (as
above).

The `.loc` attribute can be used to extract values from our `DataFrame` with
row and column _labels_:

```
# Extract the emissions (column label) of Scotland (row label)
df.loc["Scotland", "emissions"]
```

The slice operator `:` can be used to take slices of the `.loc`
attribute.

When slicing the `.loc` attribute both the start value _and_ the stop
value are returned.

This is in contrast to `:`'s behaviour with lists and NumPy arrays:

```
# Extract the population of the rows between
# and including England and Wales
df.loc["England":"Wales", "population"]
```

The `.iloc` attribute can be used to extract values from our `DataFrame` with
indices:

```
# Extract the entry in row 2 column 1
df.iloc[2, 1]
```

The slice operator `:` behaves with the `.iloc` attribute as it does with
lists, and NumPy arrays (the end point is _not_ returned):

```
# Extract df[0, 0] and df[1, 0]
df.iloc[:2, 0]
```

### 3.4.2 Appending rows

To add a new row to our `DataFrame` specify the name of our new row with the
`.loc` attribute, and assign values to it:

```
# Create a new row "Northern Ireland" from a dictionary
df.loc["Northern Ireland"] = {"population": 1910000,
                              "emissions": 20.7}
df
```

### 3.4.3 Appending columns

To append a new column to an existing `DataFrame` we specify the name of our
new column, and the values which should go in it:

```
# Create a new column "per_capita"
df["per_capita"] = df["emissions"] / df["population"]
df
```

### 3.4.4 Removing rows and columns

The Republic of Ireland is not part of the UK, so it should not appear in our
`DataFrame` of UK nations.

We can use the `.drop()` method to remove this row as:

```
# Drop the row with the "Ireland" index
df = df.drop("Ireland")
df
```

By default the `.drop()` method _assumes_ we want to remove _rows_.

To remove columns we need to explicitly tell Python to do so with the `axis`
argument:

```
# Drop the population column using the "columns" axis
df = df.drop("population", axis="columns")
df
```

### 3.4.5 Summarising DataFrames

Many of the functions from the NumPy package can be applied to Pandas
`DataFrame`s as methods. These operate on the `DataFrame` in a columnwise
manner.

For example, the column means can be computed with:

```
df.mean()
```

Additionally, we can apply these methods to individual columns by extracting
them first:

```
# Compute the mean of the emissions column
df["emissions"].mean()
```

Here's a list of some useful methods for calculating summary statistics:

| Method      | Description                                   |
| ----------- | --------------------------------------------- |
| `.head()`   | Preview the first 5 rows                      |
| `.sum()`    | Sum the values in each column                 |
| `.min()`    | Find the minimum value in each column         |
| `.max()`    | Find the maximum value each column            |
| `.mean()`   | Compute the mean of each column               |
| `.std()`    | Compute the standard deviation in each column |
| `.median()` | Compute the median of each column             |

It's worth noting that NumPy functions can also be applied *to* Pandas
`DataFrame`s:

```
import numpy as np

np.mean(df, axis=0)
```

The `axis=0` argument means we care about *column* means. To extract *row*
means we would use `axis=1`.
