### Q1e)

The `"space_heating_demand"` column records the annual space heating demand of
each archetype in _kWh_.

heating demand of each archetype in _MWh_:


```
# Your Q1e) code here
buildings["space_heating_demand_mwh"] = buildings["space_heating_demand"] / 1000
```
