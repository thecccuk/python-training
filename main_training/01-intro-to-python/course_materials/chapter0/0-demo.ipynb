Welcome to Jupyter notebooks

Jupyter notebooks are an IDE (**I**ntegrated **D**evelopment **E**nvironment),

## Hang on, there are a number of ways to write Python?

Yes, think of it like this: there are a number of different word-processing
softwares which allow us to write English prose:

  * Microsoft Word
  * Notepad
  * LibreOffice
  * Pages
  * Google Docs

All these word-processors have _different features_, but the English that we
use them to write is _fundamentally_ the same language, regardless of the
word-processor we choose.

In a similar manner, there are a number of different IDEs which allow us to
write Python code:

  * Jupyter notebooks (<- you are here)
  * Spyder
  * PyCharm
  * Notepad
  * VSCode

All theses IDEs have _different features_, but the Python which we use them to
write is _fundamentally_ the same language, regardless of the IDE we choose.

## Sweet, why Jupyter notebooks then?
