## 1.2 Assignment operations

After carrying out some computation, we often want to store the resulting
output to re-use later.

We do so with _assignment_:

````python
# Assign the value 5 to the variable x
x = 5
````

To _inspect_ the value of our variable we must type the variable name:

````python
x
````

We are free to choose variable names as we see fit (although they _cannot_
contain spaces _or_ start with a number).

However, it is _good practice_ to choose variable names which are indicative of
their contents:

````python
# Choose variable names to indicate their contents
pi = 22 / 7
pi
````

The _right hand side_ of the assignment operator can contain anything which
Python _already_ knows about.

As such we can assign variables to other variables:

````python
y = x
y
````

As we _assign_ the value on the _right-hand side_ of the equals sign, to the
variable name on the _left-hand side_ of the equals sign, we can increment the
values stored in variables as:

````python
# Increment y to y+1
y = y + 1
y
````

## 1.3 Basic data types

There are a number of _built-in_ data types of Python.

We can use the `type()` function to find out the _type_ of an object:

````python
# Hours since coffee: integer
hours_since_coffee = 1
type(hours_since_coffee)
````

Python's `int`s represent _whole numbers_.

Python has a `float` type to represent numbers which have a decimal part:

````python
# Square root of 2: float
root2 = 1.414
type(root2)
````

The `str`ing type is another built-in data type of Python.

A `str`ing is defined by surrounding characters with quotation marks (single
`'` _or_ double `"`):

````python
# Cake flavours: string
best_cake = "carrot"
type(best_cake)
````

Python supports `bool`eans, a binary logical data type.

There are two `bool`eans: `True` and `False`:

````python
# Is it sunny? True or False
is_sunny = False
type(is_sunny)
````

So far we have been introduced to four data types:

| Data type | Example 1 | Example 2           | Example 3  |
| --------- | --------- | ------------------- | ---------- |

| `str`     | `"Euler"` | `'(1 + 5**.5) / 2'` | `"cherry"` |
| `bool`    | `True`    | `False`             |            |

## 1.5 Attributes and methods

Attributes and methods are _attached_ to objects. We can think of it that

* Attributes are _variables_ which are attached to an object
* Methods are _functions_ which are attached to an object

Attributes are queried with the syntax `variable.attribute`

Methods are called with the syntax `variable.method()`

An example of a method is the `.upper()` method associated with Python's string
object:

````python
# Define a string-type variable called quiet
quiet = "bang"
quiet.upper()
````

We can store the output returned by attributes and methods:

````python
loud = quiet.upper()
loud
````

_Tab completion_ provides a really convenient way to explore the methods and
attributes associated with an object. 

Simply type the name of an object you are interested in, followed by a
full-stop, and then hit the tab key to see a list of associated attributes and
methods:

```
object.<TAB>
```

````python
# Explore the attributes and methods of the quiet variable
quiet.title()
````

## 1.6 Lists 

A list can contain a collection of objects.

Lists are constructed with square brackets `[]`. The entries of a list are
separated by commas:

````python
# List of the numbers 1, 2, 3, 4 and 5
x = [1, 2, 3, 4, 5]
x
````

````python
# Look up the type
type(x)
````

A list can contain any mix of different data types:

````python
y = [2, "be", "or", False, "2b"]
type(y)
````

We can use the `len()` function (short for length) to find out the number of
elements in our list:

````python
len(y)
````

The `sorted()` function can be used to sort list objects (amongst other
object types):

````python
y = [7, 6, 10, 8, 9]

# Create a sorted list z
z = sorted(y)
z
````

By default, `sorted()` sorts values in _ascending_ order (smallest to largest).

We can sort the values in _descending_ order (largest to smallest) if we pass
`reverse=True` to `sorted()`:

````python
y = [7, 6, 10, 8, 9]
z = sorted(y, reverse=True)
z
````

## 1.8 Extracting elements from lists

We use square brackets, `[]`, to extract elements from lists; specifying
the index of the element which we'd like

````python
# Remind ourselves what's in x
x
````

````python
# Extract the first element in x
x[0]
````

In Python `0` represents the _first_ element of a list.

So, to get the third element of `x` we would use:

````python
# Extract the third element in x
x[2]
````

In general, to extract the `n`-th element of `x` we use `x[n-1]`.

### 1.8.1 Extraction: Negative indexing

A nice consequence of indexing from `0` is the intuition of _negative
indexing_.

We may extract the _last_ element in `x` with index `-1`:

````python
# Use negative indexing to get the last entry in x
x[-1]
````

Similarly, to get the _second_-to-last value in `x`:

````python
# Extract the second-to-last value in x
x[-2]
````

The following table seeks to illustrate the relationship between positive and
negative indices for a list `[1, 2, 3, 4, 5]`:

| List           | [1,  | 2,   | 3,   | 4,   | 5]   |
| -------------- | ---- | ---- | ---- | ---- | ---- |
| Positive Index | 0    | 1    | 2    | 3    | 4    |
| Negative Index | -5   | -4   | -3   | -2   | -1   |

### 1.8.2 Extraction: Slicing

We can extract multiple values from lists with _slicing_. We do this using the
`:` operator.

When slicing lists the end point is **not** returned.

We can use slicing to extract the values from index `start` through to index
`stop-1` as `x[start:stop]`:

````python
# Extract x[1], x[2] and x[3]
x[1:4]
````

If we _don't_ specify a value for `stop` (`x[start:]`) then Python assumes we
want to slice through to the _end_ of the list:

````python
# Extract x[2], x[3] and x[4] (x[4] is the last value in x)
x[2:]
````

If we _don't_ specify a value for `start` (`x[:stop]`) then Python assumes we
want to slice from the _beginning_ of the list:

````python
# Extract x[0], x[1], and x[2]
x[:3]
````

In this manner, if we don't specify a value for `start` _or_ `stop`, we will
get back our entire list:

````python
# Returns the full list
x[:]
````
