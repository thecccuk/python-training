### Q1)

Run the following code to import the `jrpyintroduction` course package, load
the "buildings" dataset, and import the `matplotlib` package:


```
import jrpyintroduction as jrpy
import matplotlib.pyplot as plt

buildings = jrpy.data.load('buildings')
buildings = buildings.sample(200)
```

### Q1a)

Use the Pandas `DataFrame` `.plot()` method to make a scatter plot of `floor_area`
against `space_heating_co2`:


```
# Your Q1a) code here
buildings.plot('floor_area', 'space_heating_co2', kind='scatter')
```

### Q1b)

Remake your plot from above, but this time also use `plt.xlabel()` and
`plt.ylabel()` to tweak the axis labels:


```
# Your Q1b) code here
buildings.plot('floor_area', 'space_heating_co2', kind='scatter')
plt.xlabel('Floor area (m2)')
plt.ylabel('Space heating CO2 (kg)')
```

### Q1c)

Remake your plot from above, but this time change the x-axis limits to only
show floor areas between `50` and `200`:


```
# Your Q1c) code here
buildings.plot('floor_area', 'space_heating_co2', kind='scatter')
plt.xlabel('Floor area (m2)')
plt.ylabel('Space heating CO2 (kg)')
plt.xlim(50, 200)
```

### Q1d)

Remake your plot from above, but this time colour each archetype by its
`space_heating_demand`. Choose a sensible colourmap to represent these points.


```
# Your Q1d) code here
buildings.plot('floor_area', 'space_heating_co2',
               c='space_heating_demand', cmap='viridis',
               kind='scatter')
plt.xlabel('Floor area (m2)')
plt.ylabel('Space heating CO2 (kg)')
plt.xlim(50, 200)
```

### Q2b)

Run the following code (using the previously unseen `.groupby()` method). Can
you work out what we've done here? What type of object is `by_age`?


```
# Q2b) example code
by_age = buildings.groupby('age_band')['space_heating_co2'].mean()

plt.plot(by_age.index, by_age)
plt.xlabel('Age band')
```

### Q2c)

Remake the above plot, this time adding an appropriate y-axis label for the
data represented


```
# Your Q2c) code here
by_age = buildings.groupby('age_band')['space_heating_co2'].mean()

plt.plot(by_age.index, by_age)
plt.xlabel('Age band')
plt.ylabel('Mean space heating CO2')
```

### Q2d)

_Harder_

Using the `DataFrame` `.groupby()` method (similar to above), create a plot of
age band vs. _total_ space heating CO2 per age band.


```
# Your Q2d) code here
by_age = buildings.groupby('age_band')['space_heating_co2'].sum()

plt.plot(by_age.index, by_age)
plt.xlabel('Age band')
plt.ylabel('Total space heating CO2')
```
