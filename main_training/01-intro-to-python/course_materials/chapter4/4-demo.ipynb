# 4 Graphics

Many different visualisation packages have been developed for Python.

Probably the longest established of these packages is Matplotlib.

Along with its standing as a pillar of visualisation in the Python world comes
comprehensive documentation, an active development community, and a reliable
and stable plotting interface.

## 4.1 Matplotlib

Matplotlib has a number of different interfaces which we may use to construct
plots.

By far the most popular and easiest to use is the `pyplot` interface. By
convention we import this interface as `plt`:

````
import matplotlib.pyplot as plt
````

### 4.1.1 Basic plots

`plt.plot()` is a versatile function, and can be use to create line and scatter
plots.

````
temp = [12, 18, 21, 25, 30]
ice_creams = [1, 3, 9, 14, 5]

# Line plot with temperature on x axis and
# ice creams on y axis
plt.plot(temp, ice_creams)
````

Faced _only_ with the plot, how would a reader know what we're plotting?
_Temperature vs. ant infestations_? Slander, I say; we have _hardly any_ ant
infestations.

````
# Add axes labels
plt.plot(temp, ice_creams)

plt.xlabel("Temperature (degrees)")
plt.ylabel("Ice creams consumed")
````

Along with labelling our axes, we may also set the limits of the `x` and `y`
axes with the `plt.xlim()` and `plt.ylim()` functions respectively

````
# Add axes limits
plt.plot(temp, ice_creams)
plt.xlabel("Temperature (degrees)")
plt.ylabel("Ice creams consumed")

plt.xlim(10, 35)
plt.ylim(0, 14)
````

### 4.1.2 Markers, lines, and colours

We may change how our data is displayed by passing a third _optional_ argument
to `plt.plot()`

This optional parameter provides a convenient way to alter the colours,
markers, and lines used in our plot. This argument is specified as a _string_
of the form:

```
[marker][line][colour]
```

For example, to plot our data with _crosses_ coloured _red_, and a _dotted_
line, we may use:

````
plt.plot(temp, ice_creams, 'x:r')

plt.xlabel("Temperature (degrees)")
plt.ylabel("Ice creams consumed")
````

_Some_ available markers:

| Character | Description   |
| --------- | ------------- |
| `+`       | plus marker   |
| `o`       | circle marker |
| `x`       | cross marker  |
| `*`       | star marker   |
| `.`       | point marker  |

_Some_ available line-styles:

| Character | Description   |
| --------- | ------------- |
| `-`       | solid line    |
| `--`      | dashed line   |
| `-.`      | dash-dot line |
| `:`       | dotted line   |

_Some_ available colours:

| Character | Colour  |
| --------- | ------- |
| `b`       | blue    |
| `g`       | green   |
| `r`       | red     |
| `c`       | cyan    |
| `m`       | magenta |
| `y`       | yellow  |
| `k`       | black   |

Using the above tables we could create a plot using blue star markers (with no
line) as:

````
plt.plot(temp, ice_creams, '*b')

plt.xlabel("Temperature (degrees)")
plt.ylabel("Ice creams consumed")
````

## 4.2 Style sheets

We can change multiple aspects of our plot's aesthetics with _style sheets_.

You can see a list of our available style sheets with:

````
plt.style.available
````

You may also preview these styles on [this
page](https://matplotlib.org/stable/gallery/style_sheets/style_sheets_reference.html)

Once we have selected a style sheet we can apply it with the function
`plt.style.use()`.

I like the seaborn style, and so would apply it as
`plt.style.use('seaborn-v0_8')`:

````
plt.style.use('seaborn-v0_8')

plt.plot(temp, ice_creams)
plt.xlabel("Temperature (degrees)")
plt.ylabel("Ice creams consumed")
````

After a call to `plt.style.use()`, _all_ the subsequent plots we produce in the
current Python session will have the specified style applied

## 4.3 Other plot types

Line and scatter plots aren't the only type of plots which we can produce with
matplotlib. In fact, matplotlib supports a great number of different plot
types.

| Function         | Description                                            |
| ---------------- | ------------------------------------------------------ |
| `plt.hist(x)`    | Plot a histogram from count or density data            |
| `plt.bar(x, h)`  | Make a barplot with bars of height`h` at locations `x` |
| `plt.boxplot(x)` | Produce a boxplot (LQ, median, UQ, outliers) of `x`    |
| `plt.pie(x)`     | Create a pie chart from count data or percentages `x`  |

## 4.4 Plotting with DataFrames

The pandas DataFrame type has a `.plot()` method which provides a convenient
way to produce matplotlib-based plots. Let's load some data to play with:

````
# Import course package
import jrpyintroduction as jrpy

# Load buildings data
buildings = jrpy.data.load('buildings')

# Plot just a sample of the full dataset
buildings = buildings.sample(200)
````

The `.plot()` method allows a number of different plot types, which we specify
with the `kind` argument:

````
# Plot the floor_area column against the space_heating_co2 column
buildings.plot(x='floor_area', y='space_heating_co2', kind='scatter')
````

`kind` can be any of: `'line'`, `'bar'`, `'barh'`, `'hist'`, `'box'`, `'kde'`,
`'density'`, `'area'`, '`pie'`, `'scatter'` or `'hexbin'`

We can also use the `.plot()` method to colour data-points by some third
variable, passed to the `c` argument:

````
# Colour the points using the space heating demand
buildings.plot(x='floor_area', y='space_heating_co2', kind='scatter',
            c='space_heating_demand')
````

By default `.plot()` uses a rather unsatisfactory grayscale colourmap. We can
change colourmaps using the `cmap` argument.

See [this page](https://matplotlib.org/stable/tutorials/colors/colormaps) for a list
of all the available colour maps

````
# Apply our preferred colourmap
buildings.plot(x='floor_area', y='space_heating_co2', kind='scatter',
            c='space_heating_demand', cmap='viridis')
````

We can use the `alpha` argument to add some transparency to this plots points.
`alpha` is some value between 0 and 1.

````
# Add some transparency to the data points
buildings.plot(x='floor_area', y='space_heating_co2', kind='scatter',
            c='space_heating_demand', cmap='viridis', alpha=0.4)
````

## 4.5 Saving plots

We can save our plots with the `plt.savefig()` function

````
plt.plot(temp, ice_creams, '+-.g')

plt.xlabel("Temperature (degrees)")
plt.ylabel("Ice creams consumed")

plt.savefig("science.png")
````
