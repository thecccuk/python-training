## 5.2 CSV files

Comma separated value (CSV) files are a common format used to store data. In
such files data values are arranged into rows and columns, where individual
values are separated by commas.

### 5.2.1 Reading

The function `pd.read_csv()` provides a convenient way to read such data into
Python:


````
import pandas as pd

# Read in the data in the file csv_data.csv
df = pd.read_csv("csv_data.csv")
df
````

`pd.read_csv()` assumes that values are separated by commas.

However, the `sep` argument can be used if the values are separated by some
other delimiter:

```{pandas}
# Read in the data in the file tabsep_data.csv
df = pd.read_csv("tabsep_data.csv", sep="\t")
df
```

### 5.2.2 Writing

As well as _reading_ CSV files _into_ Pandas `DataFrame`s, we can also _write_
Pandas `DataFrame`s _out_ to CSV files.

````
# Write to csv using semicolon delimiter
df.to_csv("pandas_data.csv", sep=";")
````

## 5.3 Excel files

Pandas also provides functionality for reading and writing from the
ever-present and increasingly unavoidable Excel Spreadsheet.

### 5.3.1 Reading

Reading an Excel formatted file into Python is, as a Windows user, as simple
as:

```
df =  pd.read_excel("titanic.xlsx")
```

However, non-Windows users must burden themselves with another step if they
desire to read (2010) Excel files.

Namely, they must use the `openpyxl` package and read their Excel files as:

````
# Use "openpyxl" engine to load data from "titanic.xlsx" file
df = pd.read_excel("titanic.xlsx", engine="openpyxl")
````

### 5.3.2 Writing

As with `pd.read_csv()`, `pd.read_excel()` has a corresponding write method
`df.to_excel()`.


````
# Write to excel
df.to_excel("excel_sheet.xlsx")
````
