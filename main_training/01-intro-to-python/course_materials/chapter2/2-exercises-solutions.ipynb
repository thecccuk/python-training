### Q3)

The UK's projected _total_ emissions (MtCO2e) for the years 2025-2029, under the
Climate Change Committee's Balanced Pathway, are given in the following array.

_(Source: the `Economy-wide data` sheet of the [CCC Seventh Carbon Budget dataset](https://www.theccc.org.uk/publication/the-seventh-carbon-budget/#publication-downloads) - "direct emissions total" for the Balanced Pathway, rounded to the nearest MtCO2e.)_


````python
emissions = np.array([414, 396, 377, 355, 327])
````

### Q3a)

Extract the _first_ element of `emissions` (the emissions for 2025) and store it
in a variable called `starting_emissions`:


````python
# Your Q3a) code here
starting_emissions = emissions[0]
````

### Q3b)

Find the mean of the `emissions` values using the `np.mean()` function, and
store it in a variable called `average_emissions`:


````python
# Your Q3b) code here
average_emissions = np.mean(emissions)
````

### Q3c)

Work out the difference between `starting_emissions` and `average_emissions`,
and store it as `difference_to_average`:


````python
# Your Q3c) code here
difference_to_average = starting_emissions - average_emissions
````

````python
# Your Q3d) code here
conversion_factor = 1000
emissions * conversion_factor
````
