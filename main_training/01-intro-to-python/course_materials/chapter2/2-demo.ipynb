# 2 NumPy

## 2.1 But first, what is a package?

A package is a collection of Python _modules_, where a _module_ is a single
file containing Python code.

Packages, amongst other things, may define functions, variables, and classes.
Most packages you will use are hosted on [PyPI](https://pypi.org/) (pronounced
"pie pea eye").

`pip` is the preferred installer program for Python packages, and can be used
to install packages from PyPI. From the command line we install packages with:
`pip install packagename`.

To use packages we must _import_ them using the `import` keyword

## 2.2 Introduction to NumPy

[NumPy](https://numpy.org/about/) is at the _core_ of scientific computing in
Python.

Whether they know it or not, _almost every scientist_ working in Python
utilises the power of NumPy.

NumPy defines objects for storing large, multi-dimensional arrays and matrices,
and provides high-level mathematical functions to operate on these objects.

In doing so NumPy brings the computational performance of languages like
[C](https://en.wikipedia.org/wiki/C_(programming_language) and
[Fortran](https://en.wikipedia.org/wiki/Fortran) into the comfort of Python.

We can import NumPy using the `import` keyword:

```python
# Import the numpy package
import numpy
```

We use a `.` to access things from _within_ `numpy`

```python
# Access the array function from numpy (more on this later)
numpy.array([1, 2, 3])  
```

However, if we use a package a lot typing it's name over and over becomes
tedious.

We can give packages a short name by _aliasing_ them with the `as` keyword:

```python
# Alias numpy to np
import numpy as np

# Access the array function with our shorthand
np.array([1, 2, 3])
```

We can alias `numpy` to whatever shorthand we like.

However, by convention, users tend to alias `numpy` as `np`

## 2.3 NumPy arrays

NumPy's main object is the array.

NumPy's array object is _similar_ to Python's list object, but is much better
suited to data-oriented tasks.

We can construct NumPy arrays from lists:

```python
# List of the numbers 1, 2, 3, 4 and 5
y_list = [1, 2, 3, 4, 5]

# Array of the numbers 1, 2, 3, 4 and 5
y_array = np.array(y_list)
y_array
```

We can use the `type()` function to see that `y_array` really is a new data type:

```python
type(y_array)
```

### 2.3.1 Homogeneity

_All_ the entries of a NumPy array must be of _the same type_.

In this way we say that NumPy arrays are _homogeneous_.

Recall that lists can contain any mix of _different_ data types (they are
_heterogeneous_).

The array's `.dtype` _attribute_ tells us what type of data is stored within
our array:

```python
y_array.dtype
```

### 2.3.2 Elementwise operations

With lists the `+` operator performed concatenation and the `*` operator
performed repetition.

However, when used on NumPy arrays these operators perform _elementwise_
arithmetic:

```python
# Array of the numbers 1, 2 and 3
x = np.array([1, 2, 3])
x
```

```python
# Addition of arrays is performed elementwise
x + x
```

```python
# Multiplication of arrays is performed elementwise
x * x
```

NumPy's mathematical functions operate on arrays elementwise:

```python
# Exponentiate every entry in x
np.exp(x)
```

```python
# Take the (natural) logarithm of every entry in x
np.log(x)
```

```python
# Take the sine of every entry in x
np.sin(x)
```

### 2.3.3 Broadcasting

Broadcasting refers to how NumPy performs operations on arrays and scalars and
arrays of different shapes.

Subject to certain constraints, the smaller array (or scalar) is “broadcast”
across the larger array so that their shapes are compatible to perform the
desired operation: 

```python
# 100 is 'broadcast' across x to perform elementwise multiplication
x * 100
```

```python
# 100 is 'broadcast' across x to perform elementwise addition
x + 100
```

### 2.3.4 2-d arrays, and beyond!

An array's `.ndim` attribute tells us how many _dimensions_ it has:

```python
x.ndim
```

We can construct arrays with more than one dimension using lists of lists:

```python
# Constructing a 4x2 array from a list of lists
x = np.array([[1, 2],
              [3, 4],
              [5, 6],
              [7, 8]])
x.ndim
```

The `.shape` attribute of an array reports the number of entries in each of its
dimensions.

For a two-dimensional array this tells us the number of rows and columns:

```python
# (row, col)
x.shape
```

### 2.3.5 Extracting elements from arrays

We can extract elements from arrays in much the same manner as we extracted
elements from lists

```python
# Array of the numbers 1, 2, 3, 4 and 5
y = np.array([1, 2, 3, 4, 5])

# Extract the first element of y
y[0]
```

The slice operator `:` behaves as it did when indexing lists:

```python
# Extract y[1], y[2] and y[3]
y[1:4]
```

```python
# Extract the first three entries of y
y[:3]
```

With a two-dimensional array `x` we extract the entry in row `i`, column `j`
with `x[i, j]`:

```python
# Array with 4 rows and 2 columns
x = np.array([[1, 2],
              [3, 4],
              [5, 6],
              [7, 8]])

# Extract the element of x in row 0, column 1
x[0, 1]
```

```python
# Extract the element of x in row 2, column 0
x[2, 0]
```

We can use the slice operator `:` to extract entire rows and columns:

```python
# Extract all the entries in column 1 of x
x[:, 1]
```

```python
# Extract all the entries in row 2 of x (equivalent to x[2])
x[2, :]
```

Just as with lists, we can also use the slice operator to extract ranges of
values:

```python
# Extract entries in column 1 for rows 1, 2, and 3
x[1:4, 1]
```
