# Modular Code

Modularity is about two main ideas:

- Splitting your challenges into smaller components.
- Collecting together components that have related purposes

## Small, independent tasks are great!

Such tasks:

1. Are easier to describe 
2. Have fewer moving parts and less to think about
3. Have solutions that are easier to verify 
4. May arise in several projects
5. Can be worked on by other members of a team

## Before you start programming

- **Think**
 
  What are the main datasets, tasks and outputs involved here? For those tasks, which pose
  the greatest challenges (and can they be split into smaller challenges)?

- **Draw**
 
  Make a flowchart of a collection of the tasks, and connect in any datasets that are required or
  produced

- **Discuss**
 
  Has a colleague (or yourself), or a pre-existing package solved the challenges involved here?

## Don't Repeat Yourself (DRY)

We often repeat ourselves in code. Sometimes this happens by accident, sometimes it happens on
purpose.

The problems with duplicated code are that it makes code harder to read, more time consuming to
maintain, and increases the chances of bugs.

Suppose we want to plot the projected emissions for multiple sectors of the UK
economy. We'll start by loading some example data:


````
import jrpybestprac as jrpy
import pandas as pd

# Populate sector data in the current working directory
jrpy.examples.get_sectors()

# Load aviation and surface transport data into DataFrames
aviation = pd.read_csv("aviation.csv")
surface_transport = pd.read_csv("surface_transport.csv")
````

The emissions are reported in megatonnes of carbon dioxide equivalent (MtCO2eq).

Let's say we want to plot the aviation total emissions in **gigatonnes** against
year. We could do this with:


````
import matplotlib.pyplot as plt

aviation["total_emissions_gt"] = (
    aviation["total_emissions"]
    / 1000
)

plt.scatter(
    aviation["year"],
    aviation["total_emissions_gt"],
)
plt.title("Projected emissions for the aviation sector")
plt.xlabel("Year")
plt.ylabel("Total emissions [GtCO2eq]")
````

Now, what if we want to make the same plot but for the surface transport sector?

We _could_ copy-and-paste the above and change references of aviation to surface
transport:


````
surface_transport["total_emissions_gt"] = (
    surface_transport["total_emissions"]
    / 1000
)

plt.scatter(
    surface_transport["year"],
    surface_transport["total_emissions_gt"],
)
plt.title("Projected emissions for the surface transport sector")
plt.xlabel("Year")
plt.ylabel("Total emissions [GtCO2eq]")
````

In working like this, where could we make a mistake? We could:

1. Forget to change the dataset (referenced in _four_ places)
2. Forget to change the plot title
3. Duplicate a typo (e.g., `plt.xlab()` instead of `plt.xlabel()`)
4. Make plot style changes more laborious

When you find yourself copying-and-pasting code, pause and consider if there is
a way for you to avoid this duplication.

We could deduplicate the code converting emissions from Mt to Gt with a
function:


````
def convert_to_gigatonnes(data):
    data["total_emissions_gt"] = (
        data["total_emissions"] / 1000
    )
    return data

aviation = convert_to_gigatonnes(aviation)
plt.scatter(
    aviation["year"],
    aviation["total_emissions_gt"],
)
plt.title("Projected emissions for the aviation sector")
plt.xlabel("Year")
plt.ylabel("Total emissions [GtCO2eq]")
````

Further deduplication could be achieved by creating a function to capture our plotting logic:


````
def plot_total_emissions(data, sector):
    plt.scatter(
        data["year"], data["total_emissions_gt"]
    )
    plt.title(f"Projected emissions for {sector}")
    plt.xlabel("Year")
    plt.ylabel("Total emissions [GtCO2eq]")

aviation = convert_to_gigatonnes(aviation)
plot_total_emissions(aviation, "aviation")
````

Now, if we want to make the same plot for surface transport we can just call
our functions.


````
surface_transport = convert_to_gigatonnes(surface_transport)
plot_total_emissions(surface_transport, "surface transport")
````

## Writing good functions

### The docstring

The docstring:

- Defines what a function or method does

- Defines the interface
    - Inputs and input types
    - Outputs and output types
    - Argument names (few is good)

Start by deciding your function's docstring, _and then_ write its body

When thinking about your docstring, consider, eg.

- Should the function accept negative values?
- If you have written similar functions, are the arguments named consistently, do they return
  consistent data-types?
- Should the output of the function be of the same structure as the input?
- If the user passes in invalid input, should the function throw an error?

### No side-effects!

A pure function is one that has no side-effects and is self-contained. Function
definitions should be small ("small" is subjective; a reasonable upper limit is
100 lines of code in a function.)

Side-effects could include:
- writing to files
- printing to the command line
- modifying values outside of the function

If you cannot create a _succinct name_ the function might need splitting up.

A self-contained function always provides the same output for any given input.

A self-contained function is independent of:
- global variables
- random number generation
- the current time 

Suppose you have a function that reads some data and processes it:


````
def f(path):
    raw_values = read_the_data(path)
    processed_values = do(some(thing(to(values))))
    return processed_values
````

Separating the processing steps into a pure function means you aren't tied to the data-storage
format:


````
def process_the_data(x):
    processed_values = do(some(thing(to(x))))
    return processed_values
````

This makes it easier to move from a CSV file to a database to Dropbox:

```python
values = read_the_data(path)
process_the_data(values)
```

On a related note, a good function should be readable. Consider moving low-level
complexity into a separate function to assist readability.

### Generalisable functions

Consider a function that loads and cleans some data:


````
import os.path

def load_data():
    path = os.path.join("data", "health.csv")
    raw_data = read_the_data(path=path)
    return clean_the_data(raw_data)
````

Let's make this function more general by allowing the path to the data to be
passed an argument:


````
def load_data(path=os.path.join("data", "health.csv")):
    raw_data = read_the_data(path=path)
    return clean_the_data(raw_data)
````
