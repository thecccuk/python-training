### Q1a)

The following code extracts the archetype number with the highest space
heating demand in South England:


```
max_demand_id = (
    buildings
    .loc[buildings["region"] == "South England", "space_heating_demand"]
    .idxmax()
)

buildings.loc[max_demand_id, "archetype_number"]
```

What if we want to know the archetype with the highest space heating demand
in other regions?

Modularise the code above by writing a function called `get_max_demand_archetype()`.

Your function should take two arguments: `data` and `region`.


```
# Your Q1a) code here
def get_max_demand_archetype(data, region):
    max_demand_id = (
        data
        .loc[data["region"] == region, "space_heating_demand"]
        .idxmax()
    )
    return data.loc[max_demand_id, "archetype_number"]
```

### Q1c)

What if we want to know the archetype with the *lowest* space heating demand? 

Try creating a new function called `get_archetype()`, which takes three arguments:
`data`, `region`, and `demand_type` (highest/lowest).

_Hint_: you'll want to use an `if` statement in your function


```
# Your Q1c) code here
def get_archetype(data, region, demand_type):
    regional_data = data.loc[data["region"] == region]

    if demand_type == "highest":
        demand_id = regional_data["space_heating_demand"].idxmax()
    elif demand_type == "lowest":
        demand_id = regional_data["space_heating_demand"].idxmin()
 
    return regional_data.loc[demand_id, "archetype_number"]
```

### Q1d)
