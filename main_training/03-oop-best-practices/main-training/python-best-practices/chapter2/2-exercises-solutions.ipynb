# Chapter 2 Exercise solutions



### Q1)

This question will be easier to answer in VS Code or Positron.

### Q1a)

In the example Python project created during the demo, you will find a
requirements.txt with the project dependencies. Locate this file and inspect
the contents.

### Q1b)

Initialise a virtual environment using `venv` (call it whatever you like) and
activate it.

From your terminal you should have typed:

```
python -m venv venv
source venv/bin/activate
```

### Q1c)

Use `pip` to install the dependencies listed in the requirements.txt file.

From your terminal you should have typed:

```
pip install -r requirements.txt
```

### Q1d)

Try running the code in main.py. In the command line, you can run:

```
python main.py
```

If your virtual environment has the correct dependencies installed, the code
should run without error and produce an output plot called emissions.png.
