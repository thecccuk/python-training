# Reproducible Workflows

A minimal standard for data analyses is that they should be reproducible.

Organising analyses so that they are reproducible is not easy. It requires diligence and a
considerable investment of time, but _partially reproducible_ is better than not
reproducible at all.

You can think of different levels of reproducibility.

1. You can rerun the analysis next week on the _same_ computer
2. You can rerun the analysis on a _different_ computer
3. Someone else can rerun the analysis (today)
4. Someone else can rerun the analysis in 5 years time

## Organise your files

Each project should live in its own directory, which is named after the project. This directory
should contain _everything_ needed to reproduce your work, including all scripts, data,
documentation and results.

A mistake that many make is to include the full path to a file or directory on their machine:

```python
import os
import pandas as pd

project_dir = "C:/Users/bill/Documents/analysis/path/to/CoolProject"

files = {
    "cats": os.path.join(
        project_dir, "data", "cat-species.csv"
    )
}

felines = pd.read_csv(files["cats"])
```

If you move your files around, or try to share your code with a colleague those line of code won't
work.

A better solution would be to use _relative file paths_.

This means specifying paths to data _relative_ to the root of your project.

You then must run all your analysis scripts from the project root.

```python
files = {
  "cats": os.path.join("data", "cat-species.csv")
}

felines = pd.read_csv(files["cats"])
```

Now, when a colleague opens your project on their computer the relative path to that data will be
the same.

### Writing a package

The best way to collect and share functions and classes is in a Python package. 

As a developer you'll have _a lot_ of freedom to decide how best to structure your Python package.

This can be simultaneously liberating _and_ overwhelming.

To help structure your package we recommend starting from a template. Using a template:

- Avoids the paralysis-by-analysis induced by having so many structural decisions to make up-front;
- Helps newcomers familiar with the template you used, but unfamiliar with your project, get
  up-to-speed quicker;
- Doesn't waste your time reinventing the wheel;
- Allows you to concentrate on the important things (your code).

To get your package off-the-ground we recommend using the [cookiecutter tool](https://cookiecutter.readthedocs.io/en/2.0.2/)
along with [this Python Data Science template](https://drivendata.github.io/cookiecutter-data-science/):

```bash
cookiecutter https://github.com/drivendata/cookiecutter-data-science
```

This template might initialise content that your project doesn't need. Feel free to remove these if
you like: you can always add them back in later.

### The \_\_init\_\_.py file

As you investigate the example project, you will notice some
oddly named files called \_\_init\_\_.py.

Our data/ subfolder contains a preprocessing.py script with a
function called `clean_data()`. We *could* import this function using:


````
from utils.data.preprocessing import clean_data
````

Alternatively, if data/ also contains an \_\_init\_\_.py file with the following
content:

```python
from preprocessing import clean_data
```

Then we can now import `clean_data()` from our project by running:


````
from utils.data import clean_data
````

In effect, it allows us to import a function from the *subfolder* rather than
from the specific script the function belongs to. We typically only do this for
functions that are expected to be called by package users (notice that the
internal helper function `construct_data_filepath()` has been left out of
the \_\_init\_\_.py files for this particular project).

This can be very useful as we begin to split up our code across multiple
subfolders. Note that every folder, subfolder, subsubfolder, etc of a Python
package is expected to contain an \_\_init\_\_.py, even if this is empty.

### Documentation

Suppose you have written a new package or an analysis project.

You might write:

- A README.md, website, or publication to help new users get an overview of your project
- Tutorials and other long-form documentation, to help users understand how to use your code
- Help pages to explain how to use specific components of the code
- Comments to help programmers understand the inner workings of your code.

As with many things, documentation can be taken too far. Too much documentation and it may
become a maintenance burden.

### Everything with a script

Get your data in the rawest form possible.

If you need to download external data, it's best to do so via a script.

If that's not feasible, at least document the source of the data and when it was downloaded.

Write scripts for _every_ stage of data processing.

You should _never_ edit data files by hand.

## Virtual environments

It is not uncommon to encounter a situation where you are working on two
different projects, where each project requires a different version of the same
package. 

To ensure that these two projects can coexist, you'll want to be able to isolate the
environments in which they run.

We can use the `venv` tool from the command-line to create an isolated Python
environment as:

```bash
python -m venv <venv-name> 
```

Once created, your virtual environment needs to be _activated_:

```bash
source <venv-name>/bin/activate
```

Now that your virtual environment is activated, any `pip install` commands will only affect
this Python environment:

```bash
pip install pandas matplotlib
```

You can then capture these requirements in a text file with:

```bash
pip freeze > requirements.txt
```

You should regenerate the `requirements.txt` file each time you run a
`pip install` command.

A Python environment can later be restored from a `requirements.txt` file with:

```bash
pip install -r requirements.txt
```

Sharing this `requirements.txt` file with others allows them to replicate the
Python environment you used in your project.

## Version control

Do any of these file names look familiar to you?

```bash
code_v1.ipynb
code_v1b_add_plot.ipynb
code_v1b_add_plot_FINAL.ipynb
code_v1b_add_plot_FINAL_FINALv2_sept2020.ipynb
code_v1b_add_plot_FINAL_FINALv2_sept2020_Maria_edits.ipynb
```

If so, you should investigate a version control system such as git!

What are the benefits?

- Backups of your files are stored remotely using an online host.
- If you are working in a team, backups are stored in multiple locations.
- Git tracks the changes you make to files, so you have a record of what has been done, and you can
  revert to specific versions should you ever need to. This is called version control.
- Git allows changes by multiple people to be merged into a single source, thus improving
  collaborative working.

Git is a whole language in itself, and not something we can go into in this course.

However it's definitely worth considering adding version control to your toolbox. There are a great
number of resources out there on using git; we like
[this guide](https://realpython.com/python-git-github-intro/).

## Error handling

Unit testing can help to prevent _invisible_ errors (e.g., wrong output format)
from going unnoticed.

Even with unit tests, our code can still produce an error if the user calls a
function incorrectly. Consider the function below.


````
def convert_age(age_days, units="years"):
    age_multipliers = {
        "years": 1 / 365,
        "days": 1.0,
        "hours": 24.0
    }
    return age_days * age_multipliers[units]
````

This will fail if the user passes in a unit other than the three listed in
the `age_multipliers` dictionary.


````
convert_age(age_days=730, units="minutes")
````

The user will see the `KeyError` but they may not understand what they did wrong
without consulting the function source code.

Let's rewrite our function with some _error handling_:


````
def convert_age(age_days, units="years"):
    age_multipliers = {
        "years": 1 / 365,
        "days": 1.0,
        "hours": 24.0
    }
    try:
        return age_days * age_multipliers[units]
    except KeyError:
        raise ValueError(
            "'units' should be 'years', 'days' or 'hours'"
        )
````

We are now providing a more helpful error message to the user when they pass in
the wrong unit:


````
convert_age(age_days=730, units="minutes")
````

When writing functions, it's important to consider _how_ they might be
called. In doing so, we can pre-empt many user errors and even discover some
shortcomings of our code in the process.

Looking at the function again, we see that a user might accidentally pass a
string instead of a number for the `age_days` input, which would trigger a
`TypeError`.


````
def convert_age(age_days, units="years"):
    age_multipliers = {
        "years": 1 / 365,
        "days": 1.0,
        "hours": 24.0
    }
    try:
        return age_days * age_multipliers[units]
    except KeyError:
        raise ValueError(
            "'units' should be 'years', 'days' or 'hours'"
        )
    except TypeError:
        raise TypeError("'age_days' should be numeric")
````
