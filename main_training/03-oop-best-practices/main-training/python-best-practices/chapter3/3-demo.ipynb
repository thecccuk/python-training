# Working Collaboratively

One of the best ways you can improve your coding style is by learning from
other programmers.

When working in a team:

- Create an overview of your project (e.g., README.md)
- Create a shared "to-do" list (issue tracker)
- Decide on communication strategies
- Make small changes with frequent feedback

## Code Review

Code review is simply the process of reviewing code that someone else has
written.

A reviewer might:

- Make comments
- Fix issues
- Offer constructive criticism

Code reviews are beneficial because they help:

- Eliminate and catch bugs early
- Improve code readability
- Share knowledge across a team

It is easy to build code review into the merge/pull request process on git platforms such as GitLab
and GitHub.

If you're a reviewer, it's OK to:

- Ask for clarity
- Ask for the code to be made simpler
- Suggest someone, perhaps more appropriate, to conduct the review

As a reviewer:

- Explain _why_ you make any suggestions you do
- Make comments about the code _not_ the individual
- Be kind!

## Pair programming

In pair programming there is a:

- **Driver**, who writes the code and focuses solely on the task at hand.
- **Navigator**, who does not type, and so is free to spot bugs, issues, design and think about the
  wider structure of the project.

Pros of pair programming include:

- Efficient bug detection
- Producing better quality code
- Knowledge transfer
- Spending time with a colleague (coding can often be a very solitary practice!)

Of course there a few cons too. Pair programming can:

- Feel awkward
- Be tiring

This is perfectly normal! Just remember that what might take 3 hrs to code up on
your own might take only 30 minutes with an experienced colleague guiding you.

In modern programming, we often use AI copilots for these activities. More on
this later.

## Asking great questions

We all get stuck sometimes.

Being good at asking questions when we're stuck is a great skill to develop.

When asking for help, write a _reprex_ (reproducible example).

Your reprex should be simple, and minimal, and allow someone else to replicate your issue.

This allows you to share your code so other people can run it and feel your pain!

### Worked example

Suppose you have an issue with some plotting code:

```python
import datetime
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import my_constants

my_data = pd.read_csv("my_data.csv")
my_data = my_data[my_data.age > my_constants.min_age]

plot1 = plt.plot(my_data["age"], my_data["metallicity_1"])
plot2 = plt.plot(my_data["age"], my_data["metallicity_2"])
plt.legend([plot1, plot2], ["sample 1", "sample 2"])
plt.show()
```

Your code doesn't make for a good reprex as:

1. Your code doesn't run _as-is_; for someone else to run your code you'd need to share
   `my_constants.py` and `my_data.csv` with them too.
2. Modules unrelated to the issue are imported;
3. Operations unrelated to the issue are performed, e.g., filtering the data by `age`;
4. The code contains technical and potentially distracting names, e.g., `metallicity`.

Re-write as more reproducible:


````
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(
    {
        "x": [1, 2, 3],
        "y_1": [2, 3, 5],
        "y_2": [2, 2, 4],
    }
)

plot1 = plt.plot(df["x"], df["y_1"])
plot2 = plt.plot(df["x"], df["y_2"])
plt.legend([plot1, plot2], ["sample 1", "sample 2"])
plt.show()
````

You now have an example that anyone can use to reproduce the error on their
machine. This is also completely safe to share with an AI model (e.g., Copilot
or Claude) since it does not contain potentially sensitive data.

## Recap

In this chapter we have learned that:

- Code review can help us catch bugs and improve code quality.

- Pair programming is an efficient way to review code as it is being written.

- If you get stuck ask for help by writing a *reprex*.

### What about AI?

You might now be wondering how large language models (LLMs) enter into the
development workflow.

Working with an LLM is a bit like pair programming. In a short
space of time you will be able to resolve issues, learn some new techniques and
build the new feature.

However, working with an LLM can also introduce some risks:

- AI-generated code might run without error but still give you the wrong
  output.

- LLMs are unlikely to have the same contextual knowledge as someone
  else on your team. 

- LLMs can hallucinate. If your project depends on a recent package, the LLM may
  make up some function names.

- If we have a codebase that is largely made up of AI-generated code, will we
  know how to debug it when things inevitably go wrong?

With these risks in mind, here are some recommendations for responsible use of
LLMs.

**Check the code**

Read and test the code that the LLM has produced before accepting the changes.

- You should understand how your code works!
- It can also be great learning opportunity.

**Code review**

Continue to follow a standard code review process involving at least two
humans.

- If someone is on annual leave, at least one other person will be
  familiar with the code.
- No single person can be held responsible when things go wrong because of
  the code. You take joint responsibility as a _team_.

**Efficient prompting**

Provide as much context as possible to the LLM before commencing development
of the new feature.

- This will mean less time spent on frustrating back-and-forth prompts with the
  LLM.
- Fewer prompts also means lower costs.

Useful context for the LLM might include:

- What is this project for?
- What is the long term vision?
- Is there a code style template that should be followed?
- How should function docstrings be formatted?
- Which plotting theme should be used?

**Validation**

Consider writing some unit tests for your project _before_ developing any
functions.

This is a validation-centric approach to coding and it's particularly effective
in the era of LLMs:

  - By reading your unit tests, the LLM understands how you intend the code to
    be used and will adjust the functionality and input/output format
    accordingly.
  - Any AI-generated code can be quickly validated using `pytest`.
