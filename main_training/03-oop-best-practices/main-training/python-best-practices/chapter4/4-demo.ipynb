# Code Style

## What is code style?

There are surface-level qualities:

- Should you `"double-quote"` or `'single-quote'` your strings?
- Should you indent using tabs or spaces, and if you indent, how much and where should you indent?
- Should you use `camelCase` or `snake_case` for naming variables?

Then there are the structural qualities:

- How should you organise your scripts?
- Where should your functions be defined?
- Where should your documentation live?
- When there are several ways to do the same thing (e.g., for-loops,
  vectorisation, map-style functional iterators, Python's list-comprehensions)
  which should you choose?

The gold standard for Python styling is
[PEP8](https://www.python.org/dev/peps/pep-0008/). Additional style guidelines
may be defined on a project-by-project basis.

## Why even discuss style?

Code style tends to impact upon the _long-term_ success of a project or team, rather than your
productivity on any given day.

Having a defined style can make it easier to:

- Find and name variables, functions, methods and classes
- To focus on what the code _does_ rather than how it _looks_

## Automation to the rescue

Depending on the team and project, sticking to a style guide could feel like a burden.

Automation can prevent code-style from becoming a burden:

- People find it less intrusive when an automated tool tells them they've misaligned their code than
  when their colleagues do.
- Code reviews become more pleasant: you can focus on the high-level design of your code.

Python has many _linters_ and _stylers_ to help automate code styling.

Linters:

- Check your code, but _don't_ modify it
- Can identify things, such as variable names and absolute paths, that require human-input to fix
- Are highly configurable

Stylers:

- Check your code, and _do_ modify it
- Can fix things, such as white-space and quotes, that don't influence the running of your code

the capabilities of many older tools including flake8, black, isort and more.

## Examples of code styling

### Script layout

There should be a logical order to the structure of your scripts:

- All package loading code should go at the top of your scripts.

- Collect other external dependencies at the top, this includes:
    - User-defined options
    - Files
    - Databases
    - Web pages
    - Function definitions

### Organising functions

Each function must be defined before it is called. As such, it is common to put all function
definitions together in a single place towards the start of a script.

This code errors because you can't call a function before it is defined:

```
double_me(2)


def double_me(x):
    return x * 2
```

Note that functions aren't _called_ if they are used in the definition of another function.

This means the following is perfectly valid:


```
def triple_it(x):
    return double_it(x) + x


def double_it(x):
    return x * 2


triple_it(3)
```

You then _might_ choose to organise your functions in a top-down way:


```
def process_it(x):
    cleaned_up_data = clean_it(x)
    summarised_data = summarise_it(cleaned_up_data)
    return summarised_data


def clean_it(x):
    """
    How we clean the data
    """
    pass


def summarise_it(x):
    """
    How we summarise the data
    """
    pass
```

Alternatively, you may chose to order them in a bottom-up way (either is fine):


```
def summarise_it(x):
    """
    How we summarise the data
    """
    pass


def clean_it(x):
    """
    How we clean the data
    """
    pass


def process_it(x):
    cleaned_up_data = clean_it(x)
    summarised_data = summarise_it(cleaned_up_data)
    return summarised_data
```

### Meaningful names

We name:

- variables 
- functions
- scripts
- packages
- files

A _good name_ can make the intention of your code clear and easy to read.

A _bad name_ can lead to confusion, misunderstanding and bugs. You might:

- forget which of `data1`, `data2`, ... contains the data you need;
- forget that `my_plot()` can make the box plot that you are currently writing code for;
- redefine `x` in multiple places.

Acronyms and abbreviations are sometimes tempting.

However, they can be difficult to say out loud and may not be obvious to someone
who hasn't seen them before.

Functions perform an action.

As such, their names should typically start with a verb: 

- `create_graph()`
- `update_db()`
- `clean_data()` 

It's good to be _consistent_ with how you name functions across your project.

_If_ you use `load_clients()` to import the data for your clients:

- Use `load_products()`, `load_suppliers()`, ... for loading other datasets,
- Use `filter_clients()`, `save_clients()`, ...  for other functions pertaining
  to client-related datasets.

## Recap

- When writing code we have the freedom to make a lot of surface and structural
  level style choices. 

- Following a code style means you can focus on what code does rather than how
  it looks. 

- Automated tools like ruff can check certain elements of code styling for us!

- A good name should be indicative of the object, succinct and consistent with
  other object names in the project.
