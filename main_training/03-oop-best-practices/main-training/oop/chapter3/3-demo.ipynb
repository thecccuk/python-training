# Defining Classes - Instance Methods and Attributes

OOP will lead you to identify the most important kinds of object in your program and make a
computational model of them using classes.

Object-associated behaviour and information are encoded as _methods_ and _attributes_.

Here we look at classes in Python in more detail, looking at instance methods and instance
attributes.

As a class author - think about the user. They shouldn't need to know about the internals of your
class.

## The `Cohort` class

The main class that we will implement will represent results from a study carried out at different
locations around the country.

Data collected:

- heights and weights of individuals
- location of the centre where the study took place

We define a `Cohort` class, with an instance for each study cohort.


```
class Cohort():
    pass

cohort1 = Cohort()
type(cohort1)
```

The user creates a `Cohort` instance by passing in the heights, weights and centre as arguments to
the constructor `__init__()`.


```
class Cohort():
    def __init__(self, heights, weights, centre):
        pass
```

Heights / weights could be stored in different ways:

- list of (height, weight) pairs
- list of heights, and list of weights
- numpy ndarray or pandas `DataFrame` containing a height-column and a weight-column

The user shouldn't need to know how the data are stored!


```
import pandas as pd

class Cohort():
    def __init__(self, heights, weights, centre):
        self.data = pd.DataFrame({
            "heights": heights,
            "weights": weights
        })
        self.centre = centre
```

Let's create an instance of this class with some example data.


```
heights = [1.4, 1.47, 1.47, 1.58, 1.25]
weights = [32, 39, 33, 66, 24]
centre = "NCL"

my_cohort = Cohort(heights, weights, centre)
```

The `Cohort()` arguments are passed to `__init__()`.


```
# Access the data attribute
my_cohort.data
```

```
# Access the centre attribute
my_cohort.centre
```

## Instance methods - in depth

Instances of a class have their own attributes and methods.
The `self` keyword is always the first parameter for an _instance method_.


```
class Person():
    def __init__(self, name):
        self.name = name
  
    def speak(self):
        message = "Hello, I'm " + self.name + "."
        print(message)
```

```
charlie = Person("Charlie")
lucy = Person("Lucy")

charlie.speak()
lucy.speak()
```

Let's add a `.plot()` method to our `Cohort` class, so each instance can
generate a scatter plot.


```
import matplotlib.pyplot as plt

class Cohort():
    def __init__(self, heights, weights, centre):
        self.data = pd.DataFrame({
            "heights": heights,
            "weights": weights
        })
        self.centre = centre

    def plot(self):
        pandas_data = self.data
        scatter_plot = pandas_data.plot.scatter(
            x="weights", y="heights", title=self.centre
        )
        plt.show(scatter_plot)
```

```
my_cohort = Cohort(heights, weights, centre)
my_cohort.plot()
```

`self` is passed implicitly when we call `*.plot()` and refers to the object calling the method.

### Returning a value from an instance method

The `return` keyword works identically inside functions and methods: it sends results back to the
caller.

Suppose we want a method that returns the maximum height in the dataset:


```
class Cohort():
    def __init__(self, heights, weights, centre):
        self.data = pd.DataFrame({
            "heights": heights,
            "weights": weights
        })
        self.centre = centre
  
    def plot(self):
        pandas_data = self.data
        scatter_plot = pandas_data.plot(
            x="weights", y="heights", title=self.centre,
            kind="scatter"
        )
        plt.show(scatter_plot)
  
    def max_height(self):
        return self.data["heights"].max()
```

```
heights
```

```
my_cohort = Cohort(heights, weights, centre)
my_cohort.max_height()
```

### Setting an attribute using an instance method

`*.__init__()` was used to initialise attributes.
Attributes can also be modified by other instance methods.

Let's write a class that has a method for importing a dataset.
First we'll write some movies data to a `.csv` file.


```
import numpy as np

raw_movies = pd.DataFrame({
    "film": ["Toy Story", "Up", "WALL-E", "Elemental"],
    "year": [1995, 2009, 2008, np.nan]
})

raw_movies.to_csv("movies.csv", index=False)
```

For any attribute that is set by methods other than `*.__init__()`, it's good practice to declare
that attribute in the constructor (here we set the `raw_data` attribute to `None` until the dataset
is actually imported).


```
class MovieDataset():
    def __init__(self, filepath):
        self.filepath = filepath
        self.raw_data = None

    def read(self):
        self.raw_data = pd.read_csv(self.filepath)
```

Initially the `raw_data` is `None`.


```
movies = MovieDataset("movies.csv")
movies.raw_data
```

After calling `*.read()`, the `raw_data` is populated.


```
movies.read()
movies.raw_data
```

## Interlude: Mutability

Attributes can be set from:

- inside the class `self.attribute = ...`
- outside the class `object.attribute = ...`.

Many Python objects are _mutable_ in this way.
Mutability can lead to headaches

```
marcie = Person("Marcie")
marcie.name
```

```
marcie2 = marcie
marcie2.name
```

Let's change the `name` for the second variable

```
marcie2.name = "Lucy"
marcie2.name
```

That seems fine. But what's happened to the first variable?

```
marcie.name
```

The two variables point to the same object.
Any modification to an object will impact everywhere it is used.

## OOP key idea: encapsulation

Encapsulation is about two things:

- keeping things that belong together _together_
- hiding things that you don't want others to see.

Classes allow us to keep data, and the methods that act on that data together.

They also allow us to hide how we store that data, and give the user simple interfaces for accessing
it.
