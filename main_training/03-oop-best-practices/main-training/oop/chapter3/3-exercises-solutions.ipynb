### Q1)

An important concept in OOP is data-hiding. Your classes should be written such
that the user doesn't need to know how the data is stored. This is part of a
larger concept called 'encapsulation'.

The original `Cohort` class is as follows:


```
class Cohort():
    def __init__(self, heights, weights, centre):
        self.data = pd.DataFrame({
            "heights": heights,
            "weights": weights
        })
        self.centre = centre

    def plot(self):
        pandas_data = self.data
        scatter_plot = pandas_data.plot(
            x="weights", y="heights", title=self.centre,
            kind="scatter"
        )
        plt.show(scatter_plot)

    def max_height(self):
        return self.data["heights"].max()
```

Rewrite the `Cohort` class such that it stores the heights and weights as two
separate lists.

_Hint_: you will need to update the `.plot()` and `.max_height()` methods to
work with the new data format.


```
# Your Q1) code here
import pandas as pd
import matplotlib.pyplot as plt

class Cohort():
    def __init__(self, heights, weights, centre):
        self.heights = heights
        self.weights = weights
        self.centre = centre

    def plot(self):
        df = pd.DataFrame({
            "heights": self.heights,
            "weights": self.weights
        })
        scatter_plot = df.plot(
            x="weights", y="heights", title=self.centre,
            kind="scatter"
        )
        plt.show(scatter_plot)
  
    def max_height(self):
        return max(self.heights)
```

Feel free to test the new class definition with the data from the course notes:


```
heights = [1.4, 1.47, 1.47, 1.58, 1.25]
weights = [32, 39, 33, 66, 24]
centre = "NCL"
my_cohort = Cohort(heights, weights, centre)
```

```
my_cohort.max_height()
my_cohort.plot()
```

By revealing _just_ what the user needs, it is easier to reimplement a class in
the future.

### Q2)

Objects can communicate by passing messages to, or by modifying, each other.

To demonstrate this, we'll demonstrate that a `Wolf` can blow down a `House`.

### Q2a)

Define a class `House` that has an attribute `material`. The material is passed
into the constructor `__init__()` as an argument. The material will be a
string, either `"straw"`, `"sticks"` or `"bricks"` (though you needn't check
this in your code).


```
# Your Q2a) code here
class House():
    def __init__(self, material):
        self.material = material
```

You can check your code with the following code:


```
house1 = House("straw")
print(house1.material == "straw")

house2 = House("sticks")
print(house2.material == "sticks")

house3 = House("bricks")
print(house3.material == "bricks")
```

### Q2b)

Define a `Wolf` class.

This should have a method `blow_down()` that takes a `House` as it's `house`
argument (so the signature should be `blow_down(self, house)`).

If the `house` is made of `"straw"` or `"sticks"` then a wolf is able to blow it
down (the wolf should set the house material to `None` after blowing it down).

You don't have to define an `__init__()` method for `Wolf` as it doesn't need
any constructor arguments.

_Hint:_ you can update any object's attribute with `object.attribute = ...`.


```
# Your Q2b) code here
class Wolf():
    def blow_down(self, house):
        if house.material in ["straw", "sticks"]:
            house.material = None
```

You can test your class with the following code:


```
wolf = Wolf()

wolf.blow_down(house1)
print(house1.material is None)

wolf.blow_down(house2)
print(house2.material is None)

wolf.blow_down(house3)
print(house3.material == "bricks")
```

In this question we are going to develop a class for timing Python code.

### Q3a)

Python's `time.perf_counter()` can be used to obtain the current time in
seconds since some specific point in time.

Use `time.perf_counter()` to determine how long it takes Python to print
"Hello, World!" to the console.


```
import time

# Your Q4a) code here
begin = time.perf_counter()
print("Hello, World!")
end = time.perf_counter()
print(end - begin)
```

### Q3b)

Fill in the class definition below. We will use an attribute to store the
time when the user called `start()`.

In your `__init__` block, define an attribute called
`start_point` with initial value `None`.

In your `start()` method, set that attribute to the current time (in
seconds).

In your `stop()` method, use the `self.start_point` attribute to determine the
duration between when the user calls `start()` and `stop()`. The duration
should be returned using the `return` keyword.


```
class Timer():
    def __init__(self):
        self.start_point = None

    def start(self):
        self.start_point = time.perf_counter()
  
    def stop(self):
        now = time.perf_counter()
        duration = now - self.start_point
        return duration
```

The class can be tested using the following:


```
timer = Timer()
timer.start()
print("Hello, World!")
timer.stop()
```

_Tip_: users shouldn't need to know about the internal design of our class.
The `start_point` attribute is not really that useful to the user as it will
always return some arbitrary number of seconds.

A common convention is to prefix any attributes that are solely for internal use
with a `_`, so we could have named our attribute `_start_point` instead.

The user would still be able to access it with `timer._start_point`, but they
will know from the name that this is intended for internal use only.
