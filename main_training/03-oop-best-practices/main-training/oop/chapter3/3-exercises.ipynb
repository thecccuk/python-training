An important concept in OOP is data-hiding. Your classes should be written such
that the user doesn't need to know how the data is stored. This is part of a
larger concept called 'encapsulation'.

The original `Cohort` class is as follows:


```
class Cohort():
    def __init__(self, heights, weights, centre):
        self.data = pd.DataFrame({
            "heights": heights,
            "weights": weights
        })
        self.centre = centre

    def plot(self):
        pandas_data = self.data
        scatter_plot = pandas_data.plot(
            x="weights", y="heights", title=self.centre,
            kind="scatter"
        )
        plt.show(scatter_plot)

    def max_height(self):
        return self.data["heights"].max()
```
