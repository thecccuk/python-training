### Q1)

The ability to extend classes is one of the most powerful benefits of OOP.

You don't have the ability to edit classes that are defined in Python packages.
That means you can't easily add methods to the Pandas `DataFrame` class, or the
numpy `ndarray` class. So you might write stand-alone functions or add methods
to a helper class.

But, if you define a subclass, you can add methods to that subclass.

In the previous chapter we defined a class for timing code:


````
import time

class Timer():
    def __init__(self):
        self._start_point = None

    def start(self):
        self._start_point = time.perf_counter()
  
    def stop(self):
        now = time.perf_counter()
        duration = now - self._start_point
        return duration
````

Imagine you don't have the ability to edit this class, but you want to add the
ability to time the evaluation of a function to it.

That is, rather than calling


````
timer = Timer()
timer.start()
print("I'm being timed")
timer.stop()
````

... you call code that looks like this:

```python
timer = Timer()
timer.time(some_function)
```

````
def time_it(f):
    timer = Timer()
    timer.start()
    f()
    duration = timer.stop()
    return duration
````
