### Q1)

The ability to extend classes is one of the most powerful benefits of OOP.

You don't have the ability to edit classes that are defined in Python packages.
That means you can't easily add methods to the Pandas `DataFrame` class, or the
numpy `ndarray` class. So you might write stand-alone functions or add methods
to a helper class.

But, if you define a subclass, you can add methods to that subclass.

In the previous chapter we defined a class for timing code:


````
import time

class Timer():
    def __init__(self):
        self._start_point = None

    def start(self):
        self._start_point = time.perf_counter()
  
    def stop(self):
        now = time.perf_counter()
        duration = now - self._start_point
        return duration
````

Imagine you don't have the ability to edit this class, but you want to add the
ability to time the evaluation of a function to it.

That is, rather than calling


````
timer = Timer()
timer.start()
print("I'm being timed")
timer.stop()
````

... you call code that looks like this:

```python
timer = Timer()
timer.time(some_function)
```

### Q1a)

Here is a function that replicates what we are attempting to achieve:


````
def time_it(f):
    timer = Timer()
    timer.start()
    f()
    duration = timer.stop()
    return duration
````

This function accepts a function `f` that can be evaluated `f()`.

In Python you can turn any expression into a callable function using the
`lambda` keyword.


````
g = lambda: print("Hello world!")
g()
````

Use `time_it` to time how long it takes to evaluate `g`.


````
# Your Q1a) code here
time_it(g)
````

### Q1b)

Back to OOP. Define a subclass of `Timer` called `MyTimer` that has a method
`time_it` into which we can pass a function `f` to be timed.

_Hint_: You don't have to define a constructor (`__init__()` method)
for `MyTimer`


````
# Your Q1b) code here
class MyTimer(Timer):
    def time_it(self, f):
        self.start()
        f()
        duration = self.stop()
        return duration
````

### Q1c)

Use your `MyTimer` class to evaluate calculating the squares of all numbers
between 0 and 9999 using the following function.


````
squares_fn = lambda: [x**2 for x in range(10000)]
````

````
# Your Q1c) code here
my_timer = MyTimer()

my_timer.time_it(squares_fn)
````
