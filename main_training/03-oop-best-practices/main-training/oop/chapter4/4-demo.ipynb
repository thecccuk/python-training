# Object-Oriented Design - Inheritance

Inheritance provides a way to model relationships between similar types of objects.

## Modelling emissions

You are building a small model of where the UK's greenhouse gas emissions come
from.

We can think of emissions as coming from:

- Whole sectors of the economy (agriculture, aviation, etc).
- Individual dwellings.

We'll make two classes: `Dwelling` and `Sector`.

There are some common attributes:

- name
- year
- annual emissions

But we may want to include some additional attributes in the `Dwelling` class,
like:

- floor area
- how it is heated


```
class Dwelling():
    def __init__(self, name, year, emissions, floor_area, heating_system):
        self.name = name
        self.year = year
        self.emissions = emissions  # MtCO2eq per year
        self.floor_area = floor_area  # m2
        self.heating_system = heating_system

class Sector():
    def __init__(self, name, year, emissions):
        self.name = name
        self.year = year
        self.emissions = emissions  # MtCO2eq per year
```

```
scottish_bungalow = Dwelling(
    "Scottish Bungalow", 2025, 4.32, 151.54, "Gas boiler"
)
aviation = Sector(
    "Aviation", 2025, 5.852e7
)
```

## Subclasses and the class hierarchy

The two classes have a bit of duplication.

Inheritance: An "IS A" relationship

- A dwelling is not a sector
- A sector is not a dwelling
- Is there some more-general type of object, that a dwelling is and a sector is?


```
class EmissionsSource():
    # ... details omitted ...
    pass

class Dwelling(EmissionsSource):
    # ... details omitted ...
    pass

class Sector(EmissionsSource):
    # ... details omitted ...
    pass
```

Now

- each `Dwelling` is an `EmissionsSource`
- each `Sector` is an `EmissionsSource`


```
d = Dwelling()
print(isinstance(d, Dwelling))
print(isinstance(d, EmissionsSource))
print(isinstance(d, Sector))
```

That is

- `Dwelling` (and `Sector`) is a subclass of the superclass `EmissionsSource`
- `Dwelling` inherits, extends or derives from `EmissionsSource`

## The subclass constructor

All attributes and methods defined on a superclass are immediately available on it's subclasses.


```
class EmissionsSource():
    def __init__(self, name, year, emissions):
        self.name = name
        self.year = year
        self.emissions = emissions

class Dwelling(EmissionsSource):
    def __init__(
        self, name, year, emissions, floor_area, heating_system
    ):
        # Use the superclass constructor to set up common
        # attributes
        super().__init__(name, year, emissions)
        # Then add any dwelling-specific attributes
        self.floor_area = floor_area
        self.heating_system = heating_system

class Sector(EmissionsSource):
    def __init__(
        self, name, year, emissions
    ):
        super().__init__(name, year, emissions)
```

Creating objects works exactly as it did before.


```
scottish_bungalow = Dwelling(
    "Scottish Bungalow", 2025, 4.32, 151.54, "Gas boiler"
)
aviation = Sector(
    "Aviation", 2025, 5.852e7
)
```

Since `Sector` doesn't appear to have any additional attributes, we could have
just created `aviation` using the `EmissionsSource` class directly.


```
aviation = EmissionsSource("Aviation", 2025, 5.852e7)
```

But we may decide to add additional `Sector`-specific attributes later on.

## Methods and inheritance

Methods are inherited by subclasses too, so anything that makes sense for every emissions source can
be written once on the superclass.

We might want to know the total quantity of emissions over a period of time.


```
class EmissionsSource():
    def __init__(self, name, year, emissions):
        self.name = name
        self.year = year
        self.emissions = emissions

    def cumulative_emissions(self, years):
        # Emissions if this year's rate held for `years` years
        return self.emissions * years
```

Again, we have to redefine the subclasses.


```
class Dwelling(EmissionsSource):
    def __init__(
        self, name, year, emissions, floor_area, heating_system
    ):
        super().__init__(name, year, emissions)
        self.floor_area = floor_area
        self.heating_system = heating_system

class Sector(EmissionsSource):
    def __init__(
        self, name, year, emissions
    ):
        super().__init__(name, year, emissions)
```

We wrote `cumulative_emissions()` once, and it now works on `EmissionsSource` instances and on any of
it's subclasses.


```
scottish_bungalow = Dwelling(
    "Scottish Bungalow", 2025, 4.32, 151.54, "Gas boiler"
)
print(scottish_bungalow.cumulative_emissions(5))
```

```
aviation = Sector(
    "Aviation", 2025, 5.852e7
)
print(aviation.cumulative_emissions(5))
```

This is where inheritance earns its keep. Without it, a change to shared behaviour means the same
edit in every class, and a chance to get one of them wrong.

This ties in more generally with the concept of **DRY (Don't Repeat Yourself)** in
Python programming.
