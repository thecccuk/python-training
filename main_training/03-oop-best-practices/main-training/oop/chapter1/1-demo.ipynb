# Introduction to Object-Oriented Programming

## What is Object-Oriented Programming?

Object-oriented programming is a means for modelling (real-world)
entities as software objects

Every object has:
- Properties / data associated with them (**attributes**)
- Behaviours / functions they can perform (**methods**)

In Python, **everything** is an object!


```
x = [4, 2, 1, 5, 3]
x.sort() # list method
```

```
quiet = "bang"
quiet.upper() # String method
```

```
import pandas as pd


nations = {
    "nation": ["England", "Scotland", "Wales", "Ireland"],
    "population": [56490000, 5480000, 3110000, 5120000],
    "emissions": [330.5, 39.8, 34.2, 54.9]
}

df = pd.DataFrame(nations)
df.shape # DataFrame attribute
```

```
df.head() # DataFrame method
```

```
df.plot(x="population", y="emissions", kind="scatter") # DataFrame method
```

We can think of

- Attributes as **variables** belonging to an object
- Methods as **functions** belonging to an object (hence the `()`)

## Comparison to procedural programming

By *procedural* we mean a functional form of programming where we run a sequence
of functions. Consider the example below.


```
import pandas as pd

# Note: Function definitions are hidden
actor_file = "actors-2023.csv"
movies_file = "movies-2023.tsv"

# Import the data
actors_raw = pd.read_csv(actor_file)
actors_clean = clean_actors(actors_raw)

movies_raw = pd.read_csv(movies_file, sep = "\t")
movies_clean = clean_movies(movies_raw)

# Do some analysis
analyse(actors_clean, movies_clean)
```

We are loading each data set, then applying a series of functions for data
cleaning and analysis. With this approach we can end up with a lot of:

- functions that are difficult to organise.
- intermediate variables (`data_raw`, `data_clean`, etc).

Let's rephrase using an object-oriented approach:

What are the main objects?

- The **actors** and **movies** data sets

For those objects, what information do we need to store?

- Where the data is stored on disk (i.e., the file path)
- The data itself!

What behaviours can we identify?

- How to read the data into memory
- How to clean it up

The code might look something like:

We're also creating fewer variables.

### Example: plotting with Matplotlib

When learning about Matplotlib for the first time, most users start with the
`pyplot` interface.


```
import matplotlib.pyplot as plt


temp = [12, 18, 21, 25, 30]
ice_creams = [1, 3, 9, 14, 5]

plt.plot(temp, ice_creams)
plt.xlabel("Temperature")
plt.ylabel("Ice creams consumed")
plt.savefig("ice_creams.png")
```

This is procedural: it's just a series of step-by-step functions.

This approach is intuitive, but organising code for multiple plots or subplots
is hard! So more advanced programmers tend to favour the object-oriented
interface.


```
fig, ax = plt.subplots(figsize=(6, 4))

ax.plot(temp, ice_creams)

ax.set_xlabel("Temperature")
ax.set_ylabel("Ice creams consumed")

fig.savefig("ice_creams.png")
```

All operations are now methods that are attached to an object:

- `ax` for the plot.
- `fig` for the image file.
