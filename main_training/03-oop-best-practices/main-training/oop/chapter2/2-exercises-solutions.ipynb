### Q3)

The following code is broken, can you fix it?


```
class Wolf():
    def __init__(is_bad):
        self.is_bad = is_bad
    
    def greet(self):
        if is_bad:
            greeting = "I'll huff and I'll puff and I'll blow your house down"
        else:
            greeting = "Hi"
        print(greeting)
```

```
# Your Q3) code here
class Wolf():
    def __init__(self, is_bad):
        self.is_bad = is_bad

    def greet(self):
        if self.is_bad:
            greeting = "I'll huff and I'll puff and I'll blow your house down"
        else:
            greeting = "Hi"
        print(greeting)
```

Once fixed, the following code should run without any errors:


```
wolf1 = Wolf(is_bad=True)
wolf1.greet() # Should print "I'll huff and ..."
```

```
wolf2 = Wolf(is_bad=False)
wolf2.greet() # Should print "Hi"
```
