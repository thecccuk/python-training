# Objects, Classes and Instances

Object-oriented programming is built upon modelling *things* in code.

## The class is a blueprint

When building a house, you have a blueprint:

- What will the house look like?
- How will it be constructed?

In OOP we have the _class_:

- What information should the object store
- What behaviour should the object display

To define a class, we use the keyword `class`:


````
class House():
    # What information do we need to know about the house?
    pass
````

````
class Wolf():
    # What information do we need to know about the wolf?
    # What things should the wolf be able to do?
    pass
````

## The object is an instance (of a class)

Everything in Python is an _object_.

We can create objects from classes:

```python
object_name = ClassName(arguments)
```

`DataFrame` is a class in the pandas package.

With it we can create a `DataFrame` _instance_ (an object of class `DataFrame`).


````
import pandas as pd

my_data = pd.DataFrame(data={
    "heights": [1.56, 1.76, 1.44],
    "weights": [65, 68, 58]
})

my_data
````

The `type` function can tell you what kind of object you're working with:


````
print(type("abc"))
print(type(1))
print(type(my_data))
````

We can also check whether an object is of a particular class with `isinstance`:


````
print(isinstance(my_data, pd.DataFrame))

wolf = Wolf()
print(isinstance(wolf, Wolf))

print(isinstance(wolf, pd.DataFrame))
````

## Attributes store information

Attributes are values that are associated with an object.

Attributes are accessed with `object_name.attribute_name` syntax:


````
import numpy as np

# `tree_heights` is a numpy `array` instance
tree_heights = np.array([10.5, 2.6, 7.8])

# To access the `dtype` attribute:
tree_heights.dtype
````

You can actually add attributes to an object with dot-syntax (though it's uncommon):


````
house1 = House()

house1.material = "straw"

house1.material
````

Usually, the class author decides which attributes are stored for class instances.
These might be attached to the instance during it's creation.

This is possible using a method called `__init__()` which is defined _inside_
the class.


````
class Car():
    def __init__(self):
        print("Making a car")

car = Car()
````

`self` refers to the current instance of a class and is used by `__init__()`
to initialise the information / attributes stored by the object.

All cars have a model name and fuel efficiency.


````
class Car():
    def __init__(self, model, mpg):
        self.model = model
        self.mpg = mpg
````

We can now pass in values for these attributes when constructing our objects:


````
mazda = Car(model="Mazda RX4", mpg=21.0)
firebird = Car(model="Pontiac Firebird", mpg=19.2)
````

````
mazda.model
````

````
firebird.mpg
````

## Methods enable behaviour

Methods are functions that are bound to an object.

Consider the `min()` and `mean()` methods of a numpy array:


````
import numpy as np

x_list = [1, 4, 3, 5]

x_array = np.array(x_list)

print(x_array.min())
print(x_array.mean())
````

Methods are written similarly to functions, but are defined inside the class.
`__init__()`, which we just met, is a method.

Suppose we want to make a class to represent artists and have a method to indicate whether
a particular artist painted a given picture.


````
class Artist():
    def __init__(self, artist, paintings):
        self.artist = artist
        self.paintings = paintings

    def has_painted(self, painting):
        return painting in self.paintings
````

Note that we can access the current instance from a method using `self` provided it is the first
method parameter.


````
pablo = Artist("Picasso", ["Guernica", "La Vie"])
pablo.has_painted("Mona Lisa")
````
