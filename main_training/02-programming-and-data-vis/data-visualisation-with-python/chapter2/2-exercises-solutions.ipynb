The exercises use the `electricity_emissions` dataset from the demo. Run the
cell below to load it and split out the historical and Balanced Pathway series.


```
import matplotlib.pyplot as plt
import jrpyvis as jrpy

emissions = jrpy.data.load("electricity_emissions")

historical = emissions[emissions["period"] == "Historical"]
bp = emissions[emissions["period"] == "Balanced Pathway"]
```

### Q1)

Using matplotlib's object-oriented interface, create a line plot of the Balanced
Pathway `emissions` ($y$-axis) vs `year` ($x$-axis), and change the fontfamily,
fontsize and colour of the plot title and axis labels.


```
# Your Q1) code here
fig, ax = plt.subplots()
ax.plot(bp["year"], bp["emissions"])
ax.set_title(
    "UK electricity supply emissions",
    fontdict={
        "fontfamily": "monospace",
        "fontsize": 20,
        "color": "purple",
    },
)

axis_label_fontdict = {
    "fontfamily": "monospace",
    "fontsize": 18,
    "color": "green",
}

ax.set_xlabel("year", fontdict=axis_label_fontdict)
ax.set_ylabel("emissions / MtCO2e", fontdict=axis_label_fontdict)
```

### Q2a)

Change Matplotlib's runtime configuration settings so that the:

- font family is "monospace"
- fontsize is 12pt
- tick label fontsize is 10pt
- ticks face `"in"`

_Hint_: The following keys may be useful

- `font.family`
- `font.size`
- `[x|y]tick.labelsize`
- `[x|y]tick.direction`

```
# Your Q2a) code here
import matplotlib as mpl

# Using mpl.rcParams:
mpl.rcParams["font.family"] = "monospace"
mpl.rcParams["font.size"] = 12
mpl.rcParams["xtick.labelsize"] = 10
mpl.rcParams["ytick.labelsize"] = 10
mpl.rcParams["xtick.direction"] = "in"
mpl.rcParams["ytick.direction"] = "in"

# Alternatively, using mpl.rc():
mpl.rc("font", family="monospace", size=12)
mpl.rc("xtick", labelsize=10, direction="in")
mpl.rc("ytick", labelsize=10, direction="in")
```

### Q3b)

For this question we'll recreate the CCC's electricity-supply **costs** chart -
a **bar and line** figure. Run the cell below to load `electricity_costs`: the
capex, opex and net cost of the Balanced Pathway for electricity supply, in £bn
(2023 prices), taken from sheet `7.5.6` of the workbook. Reshaping to one column
per series makes it easy to plot.


```
costs = jrpy.data.load("electricity_costs")

wide = costs.pivot(index="year", columns="series", values="value")
capex = wide["Electricity supply capex"]
opex = wide["Electricity supply opex"]
net = wide["Net cost"]
years = wide.index
```

Build the figure by combining two plot types on the **same axes**. `opex` is a
*saving* (negative), so draw it first with `ax.bar(...)`: each opex bar then hangs
**down from zero**. Add `capex` as a second `ax.bar(...)`: where capex is also
negative it should stack *beneath* opex, otherwise rise from zero - pass
`bottom=opex.where(capex < 0, 0)` to do this. Finally add `net` as a **line** with
`ax.plot(...)`. All three share the same £bn axis. Add axis labels, a title and a
legend.


```
# Your Q3b) code here
fig, ax = plt.subplots()

ax.bar(years, opex, label="Electricity supply opex")
ax.bar(years, capex, bottom=opex.where(capex < 0, 0), label="Electricity supply capex")
ax.plot(years, net, color=ccc.Colors["aubergine"], label="Net cost")

ax.set_xlabel("Year")
ax.set_ylabel("Additional cost (£bn/yr)")
ax.set_title(
    "Costs and cost savings in the Balanced Pathway for electricity supply to decarbonise baseline supply"
)
ax.legend()
```

_This recreates the CCC's **Figure 7.5.6**; the original chart and its data are on
sheet `7.5.6` of the workbook._
