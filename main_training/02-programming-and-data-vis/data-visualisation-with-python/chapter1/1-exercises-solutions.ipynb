# Chapter 1 Exercise solutions







Questions 1 and 2 use the `electricity_emissions` dataset from the demo;
questions 3 and 4 use a separate `buildings` dataset, loaded further down. Run
the cell below to load the emissions data and split out the historical and
Balanced Pathway series.

_Q1 and Q2 draw on the CCC's electricity supply emissions and intensity, from
sheets `7.5.1` and `7.5.2` of the workbook (Figure 7.5.1)._


```
import matplotlib.pyplot as plt
import jrpyvis as jrpy

emissions = jrpy.data.load("electricity_emissions")

historical = emissions[emissions["period"] == "Historical"]
bp = emissions[emissions["period"] == "Balanced Pathway"]
```

### Q1a)

Using Matplotlib's **object-oriented interface**, create a line plot of the
Balanced Pathway `emissions` ($y$-axis) against `year` ($x$-axis). Add $x$ and
$y$ axis labels and a title.

_Hint_: start with `fig, ax = plt.subplots()`.


```
# Your Q1a) code here
fig, ax = plt.subplots()

ax.plot(bp["year"], bp["emissions"], label="Balanced Pathway")

ax.set_xlabel("Year")
ax.set_ylabel("Emissions / MtCO2e")
ax.set_title("UK electricity supply emissions")
```

### Q1b)

Add the `historical` emissions as a second line on the same axes, and add a
legend to distinguish the two.

_Hint_: you can add a legend with `ax.legend()`.


```
# Your Q1b) code here
ax.plot(historical["year"], historical["emissions"], label="Historical")
ax.legend()

fig
```

### Q1c)

Set the $y$-axis limits to run from $0$ to $180$.


```
# Your Q1c) code here
ax.set_ylim(0, 180)

fig
```

### Q1d)

Set the $x$-axis ticks to appear every ten years from $2010$ to $2050$, then
save your figure to a file called `electricity_emissions.png`.


```
# Your Q1d) code here
ax.set_xticks([2010, 2020, 2030, 2040, 2050])
fig.savefig("electricity_emissions.png")

fig
```

### Q2)

Create a **1 x 2 grid of subplots** showing emissions on the left and emissions
intensity on the right. Using `plt.subplots(nrows=1, ncols=2, ...)`: plot the
historical and Balanced Pathway `emissions` on the left axes and
`emissions_intensity` on the right; give each panel a title; add an $x$-axis
label ("Year") and a legend to each; and call `fig.tight_layout()` to tidy the
spacing.


```
# Your Q2) code here
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

axs[0].plot(historical["year"], historical["emissions"], label="Historical")
axs[0].plot(bp["year"], bp["emissions"], label="Balanced Pathway")
axs[0].set_title("Emissions / MtCO2e")

axs[1].plot(historical["year"], historical["emissions_intensity"], label="Historical")
axs[1].plot(bp["year"], bp["emissions_intensity"], label="Balanced Pathway")
axs[1].set_title("Emissions intensity / gCO2 per kWh")

for ax in axs:
    ax.set_xlabel("Year")
    ax.legend()

fig.tight_layout()
```

### Q3)

Questions 3 and 4 use the `buildings` dataset - the CCC's residential building
archetypes (the same data used in the Introduction course). Run the cell below to
load it and pull out the English building stock.


```
buildings = jrpy.data.load("buildings")

england = buildings[buildings["country"] == "England"]
floor_area = england["floor_area"]
space_heating_co2 = england["space_heating_co2"]
```

### Q3a)

Using the object-oriented interface, create a scatter plot of `space_heating_co2`
($y$-axis) vs `floor_area` ($x$-axis). Include axis labels.


```
# Your Q3a) code here
fig, ax = plt.subplots()

ax.scatter(floor_area, space_heating_co2)
ax.set_xlabel("Floor area / m2")
ax.set_ylabel("Space heating CO2 / kg")
```

### Q3b)

Change the size, colour and shape of the markers (the character for a diamond
marker is `"d"`).


```
# Your Q3b) code here
fig, ax = plt.subplots()

ax.scatter(floor_area, space_heating_co2, c="m", marker="d", s=10)
ax.set_xlabel("Floor area / m2")
ax.set_ylabel("Space heating CO2 / kg")
```

### Q3c)

Create a histogram of `floor_area` with 15 bins. Include axis labels and a title.


```
# Your Q3c) code here
fig, ax = plt.subplots()

ax.hist(floor_area, bins=15)
ax.set_xlabel("Floor area / m2")
ax.set_ylabel("Count")
ax.set_title("Distribution of English dwelling floor areas")
```

### Q4)

_Harder_

Create a figure of `figsize=(12, 6)` with subplots arranged in a 2 x 4 grid
(two rows, four columns). On each subplot, create a scatter plot of
`space_heating_co2` ($y$-axis) vs `floor_area` ($x$-axis) for one
`property_type`.

_Hint 1_: there are eight property types
(`buildings["property_type"].unique()`), so all eight panels are used.

_Hint 2_: `ndarray.flatten()` collapses the 2d array of axes into one dimension
to loop over, and `fig.tight_layout()` tidies the spacing.


```
# Your Q4) code here
fig, axs = plt.subplots(figsize=(12, 6), nrows=2, ncols=4)

for ax, prop in zip(axs.flatten(), buildings["property_type"].unique()):
    subset = buildings[buildings["property_type"] == prop]
    ax.scatter(subset["floor_area"], subset["space_heating_co2"], s=5)
    ax.set_xlabel("Floor area / m2")
    ax.set_ylabel("Heating CO2 / kg")
    ax.set_title(prop)

fig.tight_layout()
```
