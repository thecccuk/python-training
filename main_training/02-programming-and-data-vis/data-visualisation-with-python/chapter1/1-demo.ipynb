# 1 Plotting in Matplotlib

Matplotlib is one of the most widely used and well established visualisation
packages for Python.

Matplotlib has a number of different interfaces which we may use to
construct plots. One of the most popular of these interfaces is the
`pyplot` interface.

By convention we import this interface with the alias
`plt`.

````
import matplotlib.pyplot as plt
````

We'll start with the `electricity_emissions` dataset we met in chapter 0. Let's
load it from the course package again:

````
import jrpyvis as jrpy

emissions = jrpy.data.load("electricity_emissions")
emissions.head()
````

The `period` column splits the data into `"Historical"` values and projected
`"Balanced Pathway"` values. Let's pull out the Balanced Pathway to plot:

````
bp = emissions[emissions["period"] == "Balanced Pathway"]
````

## 1.1 Basic plotting with `pyplot`

You met the `pyplot` interface in the Introduction course, so here is a quick
reminder. A simple line plot is created with `plt.plot()`; we label the axes with
`plt.xlabel()` / `plt.ylabel()` and add a title with `plt.title()`.

````
plt.plot(bp["year"], bp["emissions"])

plt.xlabel("Year")
plt.ylabel("Emissions / MtCO2e")
plt.title("Electricity supply emissions - Balanced Pathway")

plt.savefig("emissions.png")
````

## 1.2 Object-Oriented (OO) Interface

We have seen how we can easily create plots with Matplotlib's `pyplot`
interface. However, this interface has a couple of drawbacks. For example, all
the `pyplot` functions act on the same figure. This means that if we were to
build multiple plots simultaneously, it would be difficult to keep track of
which figure we were plotting on. With Matplotlib's Object-Oriented interface,
it is much easier to keep track of what and where we are plotting.

To create a plot with the OO interface we first create an instance of the `Figure` class.

````
fig = plt.figure(figsize=(6, 4))
````

We can then add a set of axes.

````
ax = fig.add_axes(rect=[0, 0, 1, 1])
````

The argument `rect` specifies the dimensions `[left, bottom, width, height]` of
the axes. If we wish to create an `Axes` that occupies the entire `Figure` we
can also use a shorthand.

````
fig, ax = plt.subplots(figsize=(6, 4))
````

We can use the `Axes.plot()` method to create a line plot of the emissions.

````
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(bp["year"], bp["emissions"])
````

### Axis labels and limits

To add axis labels onto figures in the OO interface we use the
`Axes.set_xlabel()` and `Axes.set_ylabel()` methods.

````
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(bp["year"], bp["emissions"])

ax.set_xlabel("Year")
ax.set_ylabel("Emissions / MtCO2e")
````

We can also change the axis limits.

````
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(bp["year"], bp["emissions"])

ax.set_xlabel("Year")
ax.set_ylabel("Emissions / MtCO2e")

ax.set_xlim(2025, 2050)
ax.set_ylim(0, 35)
````

And we can add a title to our figure.

````
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(bp["year"], bp["emissions"])

ax.set_xlabel("Year")
ax.set_ylabel("Emissions / MtCO2e")

ax.set_xlim(2025, 2050)
ax.set_ylim(0, 35)

ax.set_title("UK electricity supply emissions")
````

### Tick labels

We can also control exactly where the ticks appear and how they are labelled.
Let's place a tick every five years.

````
year_ticks = [2025, 2030, 2035, 2040, 2045, 2050]
````

````
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(bp["year"], bp["emissions"])

ax.set_xlabel("Year")
ax.set_ylabel("Emissions / MtCO2e")

ax.set_xlim(2025, 2050)
ax.set_ylim(0, 35)

ax.set_title("UK electricity supply emissions")

ax.set_xticks(ticks=year_ticks)
````

### Different types of plots

So far we have only looked at line plots. However, Matplotlib allows us to
create many different types of plots, for example:

```
ax.scatter()
ax.hist()
ax.bar()
```

You'll use some of these - scatter plots and histograms - in the exercises.

### Subplots

So far we have looked at creating figures that contain only one plot. However,
often we will want to display multiple plots together to help convey more
information.

An easy way to create subplots is with the `plt.subplots()` method we have
already been using. This takes the parameters `nrows` and `ncols`.

````
f, axs = plt.subplots(
    nrows=1, ncols=2, figsize=(9, 4)
)
````

Now `axs` will be a `np.ndarray` object of `matplotlib.axes` objects:

````
type(axs)
````

````
axs.shape
````

````
type(axs[0])
````

Let's show emissions and emissions intensity side by side, indexing into `axs`:

````
f, axs = plt.subplots(
    nrows=1, ncols=2, figsize=(9, 4)
)

axs[0].plot(bp["year"], bp["emissions"])
axs[0].set_title("Emissions")
axs[0].set_ylabel("MtCO2e")

axs[1].plot(bp["year"], bp["emissions_intensity"])
axs[1].set_title("Emissions intensity")
axs[1].set_ylabel("gCO2/kWh")
````

We can also use tuple unpacking to work with the `matplotlib.axes` objects directly.

````
f, (ax0, ax1) = plt.subplots(
    nrows=1, ncols=2, figsize=(9, 4)
)

ax0.plot(bp["year"], bp["emissions"])
ax0.set_title("Emissions")
ax0.set_ylabel("MtCO2e")

ax1.plot(bp["year"], bp["emissions_intensity"])
ax1.set_title("Emissions intensity")
ax1.set_ylabel("gCO2/kWh")
````
