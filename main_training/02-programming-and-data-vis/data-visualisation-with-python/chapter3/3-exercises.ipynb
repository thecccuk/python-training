### Q2)

A grid of small multiples shows many indicators at once, but getting the layout
right - long titles, an overall title, tidy colours - is fiddly, so the code is
written for you below. **Run it** to reproduce the CCC's key-indicators figure,
then make a couple of small tweaks and re-run to see what changes:

- change the overall title text, and
- change the colour of one of the series (try another entry from `ccc.Colors`).


```
# Run this code, then make the two tweaks described above
import textwrap
import jrpyvis.ccc_theme as ccc

# The CCC theme uses constrained layout, which clashes with FacetGrid's own layout
plt.rcParams["figure.constrained_layout.use"] = False

# The two series, coloured with the CCC scenario colours from chapter 2
palette = {
    "Historical": ccc.SCENARIO_COLORS["Historical"],
    "CCC milestone": ccc.SCENARIO_COLORS["Pathway"],
}

g = sns.FacetGrid(
    data=indicators,
    col="indicator",
    col_wrap=3,
    hue="series",
    hue_order=["Historical", "CCC milestone"],
    palette=palette,
    sharey=False,
    height=2.3,
    aspect=1.3,
).map(
    sns.lineplot, "year", "value",
).add_legend()

# Shorten and wrap the long panel titles so neighbouring titles don't overlap
g.set_titles(col_template="{col_name}")
for ax in g.axes.flat:
    ax.set_xlabel("Year")
    ax.set_title(textwrap.fill(ax.get_title(), width=26))

# Leave room at the top so the overall title doesn't overlap the panels
g.figure.subplots_adjust(top=0.90, hspace=0.75)
g.figure.suptitle(
    "Key indicators for the electricity supply sector",
    y=0.985, color=ccc.Colors["vibrant purple"],
)
```
