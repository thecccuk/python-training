# 0 Introduction

Over the last few decades, Python has risen to prominence as one of the most
widely-used programming languages by statisticians and data scientists. In that
time, a plethora of popular graphics packages have been developed to tackle a
range of visualisation problems, from static 2D plots to 3D animations.

![Python plotting landscape](./plotting_landscape.png)

## 0.1 Matplotlib-related packages

[Matplotlib](https://matplotlib.org/) has been around for almost two decades and
is undoubtedly the most popular plotting library in the Python ecosystem. It is
designed primarily with static 2D plots in mind.

Matplotlib is an excellent choice whether you are a beginner or an expert with
Python:

- It boasts a reliable and stable plotting interface.
- The figure customisation options are incredibly flexible.
- Matplotlib has an active development community and comprehensive
  documentation.
- Many popular packages utilise matplotlib's functionality.
- With so many active matplotlib users, it's also very easy to get help on sites
  like [Stack Overflow](https://stackoverflow.com/) or even just by asking a
  colleague!

### Alternatives to matplotlib

For all its strengths, matplotlib _does_ have a few downsides:

- The default appearance of plots may not be particularly appealing
- Generating a publication-worthy figure can be non-trivial and require many
  lines of code.

Many alternative plotting libraries have been developed which are built
on top of matplotlib, improving on its functionality in a number of ways. Among
the most popular are:

- [*Seaborn*](https://seaborn.pydata.org/): This fantastic package allows
  complex data visualisations to be created simply and easily, whilst improving
  on the default look and feel of matplotlib figures. It also includes a large
  number of data visualisation functions, including automated regression
  fitting.

- [*Plotnine*](https://plotnine.readthedocs.io/en/stable/): A Python
  implementation of
  [*The Grammar of Graphics*](https://link.springer.com/book/10.1007/0-387-28695-0),
  which should feel instantly familiar to anyone that has used the popular R
  package [{ggplot2}](https://ggplot2.tidyverse.org/).

## 0.2 Javascript-related packages

These are used to make **interactive** 2D plots and animations. One of the most
popular examples for data science is [plotly](https://plotly.com/python/), which
is also compatible with many other languages and tools, including R, Julia and
MATLAB.

Where matplotlib produces static images, plotly renders charts you can hover,
zoom and pan - handy for exploring data or sharing a figure on the web. It has a
concise high-level interface, `plotly.express`, that plots straight from a
`DataFrame`.

Let's see it in action on some real climate data. Throughout this course we'll
recreate figures from the CCC's Seventh Carbon Budget, using
the published [Charts and data in the report](https://www.theccc.org.uk/wp-content/uploads/2025/02/The-Seventh-Carbon-Budget-Charts-and-data-in-the-report-v2.xlsx) workbook.

Here we load `electricity_emissions` - the electricity-supply emissions and
emissions intensity, taken from sheets `7.5.1` and `7.5.2` of the report - and
draw an interactive line chart (try hovering over the lines):


```
import plotly.express as px
import plotly.io as pio
import jrpyvis as jrpy

# Add the following line if plot is not visible in Jupyter
pio.renderers.default = "iframe"

emissions = jrpy.data.load("electricity_emissions")

fig = px.line(
    emissions,
    x="year",
    y="emissions",
    color="period",
    labels={"emissions": "Emissions (MtCO2e)", "year": "Year"},
    title="Electricity supply emissions - historical (2008-2023) and Balanced Pathway (2025-2050)",
)
fig.show()
```

_This recreates the CCC's **Figure 7.5.1**; the original chart and its data are
on sheet `7.5.1` of the workbook._

The rest of this course focuses on **static** figures, but it's worth knowing
plotly is there when interactivity helps.

## 0.3 Other options

- *OpenGL-related packages*: These are a great option if you wish to produce
  interactive 3D figures. Some examples include
  [PyOpenGL](http://pyopengl.sourceforge.net/),
  [visvis](https://pypi.org/project/visvis/) and [pyglet](https://pyglet.org/).

- *D3.js-related packages*: These can be used to create interactive data
  visualisations which are optimised for the web, from declarative JSON
  specifications. Examples include [mpld3](https://mpld3.github.io/) (which
  brings together D3.js and matplotlib) and
  [Vega-Altair](https://altair-viz.github.io/).
