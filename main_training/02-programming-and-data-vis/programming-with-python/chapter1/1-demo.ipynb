# Loops

In programming, it is common for us to wish to perform some tasks repeatedly.
We can accomplish such tasks with _loops_.

For example, we may wish to:

  * Perform the same analysis on multiple different datasets
  * Perform some operation on each row of a DataFrame
  * Load all the files in a folder with a particular file extension

Here we will see three different ways of constructing loops in Python.

## For loops

We can use for loops to iterate over objects such as lists, strings, and
arrays.

The key elements in a Python for loop are:

* _The keyword `for`_: we use this to tell Python that we are about to write a
  `for` loop

* _The loop variable_: this will be the value that changes at each iteration
  of our loop

* _The `in` keyword_: this separates our loop variable from our iterable

* _The iterable_: this is the object that we shall "iterate" over. It is
  often a list, string, or generator (more on these later)

* _The colon `:`_: this marks the end of the for loop header

* _The loop body_: this is the code that is executed at every iteration of our
  loop. Our loop body must be indented by 4 spaces

## For loop example

We can construct a simple for loop as:


```
for i in [5, 6, 7, 8]:
    print(i)
```

Here our loop variable was `i` and our iterable was the list `[5, 6, 7, 8]`.

Our loop body (the code to be executed at each iteration of our loop) contained
only the code `print(i)`.

The above loop is equivalent to the more tediously typed:

```
print(5)
print(6)
print(7)
print(8)
```

Here we only looped over a list with 4 values. But what if we wanted to loop
over lots more values? Would we have to type all these values out individually?

Fortunately not. Python's `range()` construct can help us here:

```
for j in range(10):
    print(j)
```

Here our loop variable was `j` and our iterable was `range(10)`.

Again, our loop body contained only the code `print(j)`.

The above loop is equivalent to the rather cumbersome:

```
print(0)
print(1)
print(2)
print(3)
print(4)
print(5)
print(6)
print(7)
print(8)
print(9)
```

`range()` has the same `start`, `stop`, `step` pattern that we saw with list
slicing:

```
for x in range(1, 10, 2):
    print(x)
```

Here our loop variable was `x` and our iterable was `range(1, 10, 2)`.

This loop was then equivalent to the following code:

```
print(1)
print(3)
print(5)
print(7)
print(9)
```

## While loops

The `while` loop provides another way to construct loops in Python.

With the `while` loop we execute a set of statements so long as a condition is
`True`.

The key elements in a Python while loop are:

* _The keyword `while`_: we use this to tell Python that we are about to write a
  `while` loop

* _The condition_: something that evaluates to `True` or `False`

* _The colon `:`_: this marks the end of the `while` loop header

* _The loop body_: this is the code that is executed at every iteration of our
  loop. Our loop body must be indented by 4 spaces

### While loop example

We can construct a simple `while` loop as:

```
i = 0

while i < 10:
    print(i)
    i = i + 1
```

In this example our condition was `i < 10`, and our loop body printed and then
incremented `i`.

Let's consider another example, in which we print all the `i`s that when
squared, are less than 70:

```
i = 0

while i**2 < 70:
    print(i)
    i = i + 1
```

### Aside: the `+=` syntax

Python has a nifty shorthand for performing operations like `i = i + 1`, and
that is `i += 1`

We could use this in our `while` loop examples as:

```
i = 0

while i < 10:
    print(i)
    i += 1
```

## Enumeration

Sometimes we may wish to loop over some values, and have a counter that
represents our progress through these values.

`enumerate()` constructs an iterable that has this behaviour:


```
seasons = ["Spring", "Summer", "Autumn", "Winter"]
for count, value in enumerate(seasons):
    print(count, value)
```

## Zipping

We can use `zip()` to loop over two (or more) iterables at once:

```
first_names = ["Regina", "Cady", "Ms."]
last_names = ["George", "Heron", "Norbury"]

for first, last in zip(first_names, last_names):
    # Use Python's F-strings to join the loop variables with a space in between
    print(f"{first} {last}")
```
