### Q4d)

_Harder_

The (previously unseen) `pd.concat()` function can be used to join a list of
`DataFrame`s together. The syntax for this is:

```
pd.concat([df1, df2])
```

where `df1` and `df2` are the `DataFrame`s to be joined.

Each file holds the emissions for a single sector, but the sector is only
recorded in the _filename_, not inside the file. Read in "aviation.csv" and
"surface_transport.csv", add a `sector` column to each recording which sector
it is, and join them together with `pd.concat()`.


````
# Your Q4d) code here
import pandas as pd

aviation = pd.read_csv("aviation.csv")
aviation["sector"] = "aviation"

surface_transport = pd.read_csv("surface_transport.csv")
surface_transport["sector"] = "surface_transport"

pd.concat([aviation, surface_transport])
````

### Q4e)

_Harderer_

Starting with an empty `DataFrame` (`df = pd.DataFrame()`), loop over the
files in your current directory. For each file that is a `csv` file, read it
in, add a `sector` column recording the sector (taken from the filename), and
use `pd.concat()` to append it to `df`.

_Hint_: this time you don't know the sector names up front, so take each one
from its filename by dropping the extension, eg. `file.replace(".csv", "")`,
and add it with `sector_data["sector"] = ...`.


````
# Your Q4e) code here
df = pd.DataFrame()
for file in files:
    if file.endswith(".csv"):
        sector_data = pd.read_csv(file)
        sector_data["sector"] = file.replace(".csv", "")
        df = pd.concat([df, sector_data])

df
````

I'm trying to use `enumerate()` in a for loop to create a list: `['e', 'n', 'u', 'm', 'e', 'r', 'a', 't', 'e']`

However, the code isn't doing what I want. Please could you take a look and fix
it?

````
string = "enumerate"
# Create list to store letters in
output = [0] * len(string)

# Fixed code :)!
for i, letter in enumerate(string):
    output[i] = letter
````
