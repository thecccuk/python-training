# Comprehensions

Comprehensions are great for building new data structures quickly and with
minimal code.

## List comprehensions

List comprehensions provide a convenient way of quickly and efficiently
building lists.

### Multiplying values in a list

Let's imagine that we have some list `x = [1, 2, 3, 4, 5]`, and that we
wish to multiply each entry in `x` by `2` _without_ using any additional
packages.

We _could_ solve this problem with a `for` loop:

````
x = [1, 2, 3, 4, 5]

# I'm an empty list
y = []

for i in x:
    # Append 2 * i to our list y
    y.append(2 * i)

y
````

Alternatively, we could have solved this problem with a list comprehension:


````
z = [2 * i for i in x]

z
````

The syntax of the list comprehension follows the pattern:

```
[do something for value in iterable]
```

### Uppercasing strings in a list

Suppose we have a list of strings where each string is in lowercase. We would
like to create a new list, with the strings from the original list uppercased.

This problem could be solved with a `for` loop:


````
pudding = ["how", "can", "you", "have", "any", "pudding", "if", "you",
           "don't", "eat", "your", "meat?"]

# Empty list to append to
PUDDING = []

for word in pudding:
    PUDDING.append(word.upper())

PUDDING
````

Alternatively, using our new-found comprehension skills, we could have solved
this problem as:


````
#         [do something for value in iterable]
PUDDING = [word.upper() for word in pudding]

PUDDING
````

### `if` statements _within_ comprehensions

For the examples in this section we will populate the `chapter3/` directory
with the per-sector emissions files we met in the previous chapters:


````
# Run this chunk to populate example data files to play with
import os
import jrpyprogramming as jrpy

# Ensure we are in the chapter3 directory
os.chdir("../chapter3")

jrpy.data.populate_sectors()
````

We could use the `os.listdir()` function to list all the files in our current
directory:


````
# Get list of files in current directory
files = os.listdir()

files
````

We could then use an `if`-statement _within_ a `for`-loop to create a list of
all the `.csv` files in our current directory:


````
# Empty list to append csv filenames to
csv_files = []

# Loop over names of files in current directory
for file in files:
    # Capture filename if it has a .csv extension
    if file.endswith(".csv"):
        csv_files.append(file)

csv_files
````

We could also solve this problem with a list comprehension by noting that the
simpler comprehension pattern seen previously can be extended to:

```
[do something for value in iterable if condition]
```

To create our list `csv_files` we could use this pattern as:


````
csv_files = [file for file in files if file.endswith(".csv")]

csv_files
````

## Dictionary comprehensions

Dictionary comprehensions are _similar_ to list comprehensions: providing a
concise syntax to quickly construct dictionaries.

Let's continue our `.csv` file example from above, but this time let's create a
dictionary of key-value pairs that map from filename to pandas `DataFrame`:


````
# Load the pandas package
import pandas as pd

# Create an empty dictionary
data = {}

# Loop over names of csv files in current directory
for csv_file in csv_files:
    data[csv_file] = pd.read_csv(csv_file)

data
````

Alternatively, we could have used a dictionary comprehension to solve this
problem as:


````
data = {csv_file: pd.read_csv(csv_file) for csv_file in csv_files}

data
````

A dictionary comprehension adheres to the following pattern:

```
{key: do something for value in iterable}
```

Much in the same way as with list comprehensions, conditionals can also be
included in dictionary comprehensions. This lets us filter to the `.csv` files
directly, without building `csv_files` first:


````
data = {file: pd.read_csv(file) for file in files if file.endswith(".csv")}

data
````

## Lists and dictionaries: which and when?

We've now used both _lists_ and _dictionaries_ to hold our data, so it's worth
pausing on when to reach for each.

* A **list** is an _ordered_ collection. Reach for a list when the order
  matters and you access items by their position, or you simply loop over them
  in turn - like our `csv_files`, or the years in a time series.

* A **dictionary** is a collection of _key-value pairs_. Reach for a dictionary
  when you want to look something up by a meaningful _label_ rather than a
  position - like our `data`, where we look up a `DataFrame` by its filename,
  or an emissions figure by its sector name.

As a rule of thumb: if you find yourself trying to remember "which position was
that in again?", a dictionary is probably the better fit.
