### Q1)

Suppose we have a list of numbers: `numbers = [1, 24, 35, 67, 81, 2, 45, 89, 12]`,
and we want to construct a new list `odd_numbers`, containing (you guessed it)
only the _odd_ entries of `numbers`.

### Q1a)

Use an `if`-statement _within_ a `for`-loop to create a list `odd_numbers`,
containing only the odd entries from the list `numbers`.

_Hint_: An integer `x` is odd if `x % 2 == 1`.


```
# Your Q1a) code here
numbers = [1, 24, 35, 67, 81, 2, 45, 89, 12]

odd_numbers = []
for number in numbers:
    # If a number has remainder 1 on division by 2 then it must be odd
    if number % 2 == 1:
        odd_numbers.append(number)

odd_numbers
```

### Q1b)

Use an `if`-statement _within_ a list comprehension to create the `odd_numbers`
list, as above:


```
# Your Q1b) code here
numbers = [1, 24, 35, 67, 81, 2, 45, 89, 12]

odd_numbers = [number for number in numbers if number % 2 == 1]
```

### Q2)

For this question, run the following code to load the
"energy_demand" dataset. It's a subset of the Climate Change Committee's
[Seventh Carbon Budget dataset](https://www.theccc.org.uk/publication/the-seventh-carbon-budget/#publication-downloads),
taken from the "Economy-wide data" sheet, and holds the UK's projected primary
energy demand (TWh) by source and year, under the Balanced Pathway:

```
import jrpyprogramming as jrpy

energy_demand = jrpy.data.load("energy_demand")

# Use the .head() method to preview the data
energy_demand.head()
```

### Q2a)

Run the following code, what does it do?

```
energy_demand["source"].unique()
# Grabs the unique values within the source column
```

### Q2b)

The following function can be used to compute the mean of the
'demand' column for a given energy source:

```
def compute_mean_demand(x, source):
    data_subset = x[x["source"] == source]
    return data_subset["demand"].mean()
```

With this, `compute_mean_demand(energy_demand, "wind")`
can be used to compute the mean demand for the `"wind"`
source.

Use a `for` loop to loop over `energy_demand["source"].unique()`
and append the mean value of each source's 'demand' to
a list called `mean_demands`.


```
# Your Q2b) code here
# Create an empty list to store results in
mean_demands = []
for source in energy_demand["source"].unique():
    mean_demand = compute_mean_demand(energy_demand, source)
    mean_demands.append(mean_demand)

mean_demands
```

### Q2c)

Rewrite your `for` loop from Q2b) as a list comprehension:


```
# Your Q2c) code here
mean_demands = [
    compute_mean_demand(energy_demand, source)
    for source in energy_demand["source"].unique()
]
mean_demands
```

### Q2d)

_Harder_

Use a dictionary comprehension to create a dictionary which
maps the name of the source to its mean demand.


```
# Your Q2d) code here
mean_demands = {
    source: compute_mean_demand(energy_demand, source)
    for source in energy_demand["source"].unique()
}
mean_demands
```

### Q3)

_Harderer_

The Climate Change Committee projects the UK's primary energy demand from each
source in 2025. Let's suppose someone handed us a selection of these figures as
a Python dictionary...


```
source_demand = {
    "oil": 755.9,
    "gas": 720.1,
    "wind": 244.6,
    "bioenergy": 161.0,
    "other fuels": 2.9
}
```

These figures are in terawatt-hours (TWh). Suppose we'd prefer them in
_gigawatt-hours_ (GWh) instead.

### Q3a)

Use a for loop to create a dictionary giving each source's demand in GWh:

_Hint 0_: There are `1000` gigawatt-hours in `1` terawatt-hour.

_Hint_ 1: Python dictionaries have a method `.items()` which gives us a way to
loop over their keys and values simultaneously


```
# Your Q3a) code here

# Create an empty dictionary
demand_gwh = {}

# Loop over key-value pairs in source_demand with the .items() method
for source, demand in source_demand.items():
    # Create a key-value pair
    demand_gwh[source] = demand * 1000

demand_gwh
```

### Q3b)

Create the `demand_gwh` dictionary again, but this time use a
dictionary-comprehension, rather than a for-loop:


```
# Your Q3b) code here

demand_gwh = {source: demand * 1000
              for source, demand in source_demand.items()}

demand_gwh
```

### Q3c)

"other fuels" is a catch-all bucket rather than a specific energy source, so
we'd rather report it separately and not lump it in with the named sources.

Add an `if`-statement to your for loop from Q3a) to ensure only the sources
which _aren't_ "other fuels" are converted from TWh to GWh:


```
# Your Q3c) code here

# Create an empty dictionary
demand_gwh = {}
# "other fuels" is reported separately, so leave it as-is
demand_gwh["other fuels"] = source_demand["other fuels"]

for source, demand in source_demand.items():
    # "other fuels" is a catch-all, so don't convert it
    if source != "other fuels":
        demand_gwh[source] = demand * 1000

demand_gwh
```

### Q3d)

Create the `demand_gwh` dictionary as in Q3c) again, but this time use an
`if`-statement within a dictionary-comprehension:


```
# Your Q3d) code here

demand_gwh = {source: demand * 1000
              for source, demand in source_demand.items()
              if source != "other fuels"}

demand_gwh["other fuels"] = source_demand["other fuels"]

demand_gwh
```
