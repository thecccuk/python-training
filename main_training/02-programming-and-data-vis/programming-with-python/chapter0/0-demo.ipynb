# Conditionals and Booleans

When programming we often want to _conditionally_ execute certain code.

In this way we will be able change a program's behaviour and output depending
on the outcome of some tests.

For example, we may wish to:

* Compute the square root of a number, _if_ that number is positive

* Save data to a file, _if_ that file doesn't already exist

* Extract a row from a DataFrame, _if_ the row has some certain quality

## Simple conditions

Before we can write programs that conditionally execute code, we first need
to learn how to write conditions (our tests).

Whenever we ask Python a question (our conditions / tests), Python will always
respond with either a `True` or a `False`.

We can check whether two values are _equal_ with an equality test: `==`

```
x = 5

# Does x equal 5?
x == 5
```

```
x = 5
y = 7

# Does x equal y?
x == y
```

To test if two values are _not_ equal, we use the `!=` operator

```
# Does x not equal y?
x != y
```

We can also test for _inequality_ with the greater than and less than symbols:

```
x = 5
y = 7

# Is x less than y?
x < y
```

```
x = 5
y = 7

# Is x less than or equal to y?
x <= y
```

```
x = 5
y = 7

# Is x greater than y?
x > y
```

```
x = 5
y = 7

# Is x greater than or equal to y?
x >= y
```

## Multiple conditions

Quite often we'll want to test _more than one_ condition.

This is possible with the introduction of two new operators: `&` and `|`.

If we put an ampersand (`&`) between two conditions, we are testing if _both_
conditions are `True`.

```
x = 5
y = 7

# Is both x greater than 0 *and* y greater than 10?
(x > 0) & (y > 10)
```

When combining multiple tests with `&` and `|` we isolate the individual tests
with parentheses:

```
x = 5
y = 7

# Is both x greater than 0 *and* y less than 10?
(x > 0) & (y < 10)
```

If we put a pipe (`|`) between two conditions, we are testing if _either_ of
the conditions are `True`:

```
# Is either x greater than 0 *or* y greater than 10?
(x > 0) | (y > 10)
```

## `if` statements

The `if` statement allows us to conditionally execute code.

The key elements in a Python `if` statement are:

* _The keyword `if`_: we use this to tell Python that we are about to create an
  `if` statement

* _A condition_: this is a test that will evaluate to `True` or `False`

* _The colon `:`_: this marks the end of the if header 

* _The conditional code_: this is the code that is executed when our condition
  evaluates to `True`. _The conditional code body must be indented by 4 spaces_.

## Simple `if` statements

Let's create a simple `if` statement, using the condition `temp >= 18`:


```
temp = 20

if temp >= 18:
    print("Don the shorts.")
```

```
temp = 4

if temp >= 18:
    print("Don the shorts.")
```

We can also use conditions with `&` and `|` in our `if` statements:

```
temp = 20
day = "Saturday"

if (temp >= 18) & (day == "Saturday"):
    print("Prepare the BBQ.")
```

## The `else` statement

An `else` statement can be used to contain a block of code that should be
executed if the `if` conditional evaluated to `False`.

Again, as with `if`, the code associate with the `else` statement should be
indented by four spaces:

```
temp = 20

if temp >= 18:
    print("Don the shorts.")
else:
    print("Maybe wear something full-length...")
```

```
temp = 4

if temp >= 18:
    print("Don the shorts.")
else:
    print("Maybe wear something full-length...")
```

## The `elif` statement

The `elif` statement allows us to combine multiple checks one after another.

Once an expression has been evaluated to `True` Python will evaluate that code
block and skip to the end:


```
temp = 16

if temp >= 18:
    print("Don the shorts.")
elif temp >= 14:
    print("Are three-quarters still in fashion?")
else:
    print("Maybe wear something full-length...")
```

We can chain as many `elif` statements together as we like:

Remember: once an expression has been evaluated to `True` Python will evaluate
that code and skip to the end:


```
x = -1

if x < -10:
    print("x is teeny-tiny")
elif x < 0:
    print("x is between -10 and 0")
elif x < 10:
    print("x is between 0 and 10")
else:
    print("oh lawd he comin'")
```

## Boolean indexing

We can use booleans (`True` and `False`) to extract data from `DataFrame`s.

This idea becomes particularly powerful when we derive our booleans from a
conditional test.

In this way we will be able to _conditionally_ extract data from a `DataFrame`.

Let's create a pandas DataFrame from a dictionary to play with. Here we have
the populations and territorial greenhouse-gas emissions (in MtCO2e) of four
nations:

```
import pandas as pd

nations = {
  "nation": ["England", "Scotland", "Wales", "Ireland"],
  "population": [56490000, 5480000, 3110000, 5120000],
  "emissions": [330.5, 39.8, 34.2, 54.9]
}

nations = pd.DataFrame(nations)
nations
```

We can then extract values from this with lists, arrays, or Series' of booleans:

```
mask = [True, False, False, True]

nations.loc[mask]
```

This has returned rows `0` and `3`; the positions in `mask` containing `True`.

So far, so not that useful...

However, this idea becomes more powerful when we construct our `True` and
`False` values from the output of a test:

```
# Construct boolean mask
mask = nations["population"] > 5000000
mask
```

We can then use `mask` to extract all the rows of `nations` with
`"population"`s greater than `5000000`:

```
nations.loc[mask]
```

Here, we could have achieved the above two step process in a single line of
code:

```
# Extract rows of nations with population > 5000000
nations.loc[nations["population"] > 5000000]
```

We can perform more complex conditional extracting with `&` and `|`:

```
mask = (nations["population"] > 5000000) & (nations["emissions"] < 100)
nations.loc[mask]
```
