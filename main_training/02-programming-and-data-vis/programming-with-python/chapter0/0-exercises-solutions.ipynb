### Q2a)

Write a simple `if` statement which computes `x = x ** .5` if `x` is greater than

Check that your statements behave as expected.

```
# Your Q2b) code here
if x >= 0:
    x = x ** .5
else:
    print("cannot compute sqrt")
```

### Q2c)

Add an `elif` statement to the construct above to set `x = "i"` if `x` is

```
# Your Q2c) code here
if x >= 0:
    x = x ** .5
elif x == -1:
    x = "i"
else:
    print("cannot compute sqrt")
```

### Q4)

In this example we will make use of the buildings dataset, which describes the
UK's residential building stock as a set of "archetypes". It's the same dataset
you'll have seen in the Introduction to Python course. Let's start by loading
it from the course package:

```
import jrpyprogramming as jrpy

buildings = jrpy.data.load('buildings')
buildings.head()
```

### Q4a)

Let's look for a home that's cheap to keep warm.

Extract all the archetypes which have an annual `space_heating_demand` _less
than_ 3000 (kWh).


```
# Your Q4a) code here
buildings.loc[buildings["space_heating_demand"] < 3000]
```

### Q4b)

Okay, great, we can extract homes that are cheap to heat. _But_, I'd also like
somewhere compact, so I don't want a `floor_area` of 50 (square
metres) or more.

Extract all the archetypes with a `space_heating_demand` less than 3000, _and_
a `floor_area` less than 50.


```
# Your Q4b) code here
buildings.loc[(buildings["space_heating_demand"] < 3000) & (buildings["floor_area"] < 50)]
```

### Q4c)

Given my fussiness, how many archetypes do we have left to choose from?


```
# Your Q4c) code here
buildings.loc[(buildings["space_heating_demand"] < 3000) & (buildings["floor_area"] < 50)].shape[0]
```
