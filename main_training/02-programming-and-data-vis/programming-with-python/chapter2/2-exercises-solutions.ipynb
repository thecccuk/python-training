### Q4a)

Write a function called `combine_sectors` which takes one argument, `files` (a
list of filenames). The function should loop over `files`, and for each `csv`
file: read it in, add a `sector` column taken from the filename, and
concatenate it onto a combined `DataFrame`. The function should `return` the
combined `DataFrame`.

_Hint_: this is your answer to chapter 1's Q4e), wrapped in a `def`.


```
# Your Q4a) code here
def combine_sectors(files):
    df = pd.DataFrame()
    for file in files:
        if file.endswith(".csv"):
            sector_data = pd.read_csv(file)
            sector_data["sector"] = file.replace(".csv", "")
            df = pd.concat([df, sector_data])
    return df
```

### Q4c)

Add a docstring to your `combine_sectors` function describing what it does.


```
# Your Q4c) code here
def combine_sectors(files):
    """
    Combine per-sector emissions csv files into a single DataFrame.

    Parameters
    ----------
    files : list of str
        Filenames to combine. Non-csv files are ignored.

    Returns
    -------
    pandas.DataFrame
        The concatenated data, with a `sector` column taken from each
        filename.
    """
    df = pd.DataFrame()
    for file in files:
        if file.endswith(".csv"):
            sector_data = pd.read_csv(file)
            sector_data["sector"] = file.replace(".csv", "")
            df = pd.concat([df, sector_data])
    return df
```
