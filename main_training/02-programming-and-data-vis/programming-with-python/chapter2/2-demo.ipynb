# Functions

A _function_ is a re-usable block of code that is only run when we call it.

Functions are a really great way to encapsulate tasks that we perform
regularly.

As our programs get longer and become more complex, using functions can help
keep our code tidy and maintainable.

## Defining functions

The key elements in a Python function definition are:

* _The keyword `def`_: we use this to tell Python that we are about to `def`ine a
  function

* _The function name_: this allows us to refer back to our function later

* _Parentheses_: this is where we specify the arguments (or lack thereof) that
  our function can expect to receive

* _The colon `:`_: this marks the end of the function header

* _The function body_: this is the code that is executed whenever we call our
  function. Our function body must be indented by 4 spaces

## Simple functions

Let's see a simple function definition:

```
def hello(person):
    print("Hello " + person + ", it's me: your computer.")
```

We can _call_ this function by it's name: `hello`.

To call this function, we must pass through a value to be interpreted as the `person`
variable:

```
hello("Paul")
```

```
hello("John")
```

## Returning values

Our functions may optionally _return_ values. We do so with the `return`
keyword.

Let's define a new function called `compute_square`, which takes a single
argument and `return`s the square of the argument passed:

```
# Define a function with a return statement
def compute_square(x):
    return x ** 2
```

We can then call our function as:

```
compute_square(12)
```

As our function is now `return`ing a value, we can assign the `return`ed value
to a variable for re-use later:

```
squared = compute_square(7)
squared
```

### Returning multiple values

Our functions may also return _multiple_ values. We specify all the objects
we'd like to return, separated by commas, on the return line:

```
def test_cases(string):
    return string.lower(), string.upper(), string.title()
```

Our function will now return a _tuple_ with three entries.

Although we've not been formally introduced to the tuple before, it is similar
to the familiar list object.

The difference is that the tuple is _immutable_ (lists are _mutable_) and so we
_can't_ change the values stored in a tuple, unlike how we can with lists.

```
cases = test_cases("All type no face")
type(cases)
```

We can extract values from our tuple just as we can with a list:

```
print(cases[0])
print(cases[1])
print(cases[2])
```

We can also "unpack" these values when we call our function:

```
lower, upper, title = test_cases("hello, goodbye")
print(lower)
print(upper)
print(title)
```

## Multiple Arguments

Our functions can take _multiple_ arguments. When we specify multiple arguments
we must separate them by commas:

```
def divider(x, y):
    return x / y
```

We can call such as function as:

```
divider(22, 7)
```

The _order_ in which we pass through our values is important.

The first value that we pass to our function `divider()` will be interpreted
as the argument `x`.

The second value that we pass will be interpreted by our function as the
argument `y`:

```
divider(7, 22)
```

Rather than implicitly relying on the order of the arguments, we may sometimes
explicitly specify which values would be interpreted as which variables.

In doing so we can pass our values through in any order we'd like. Consider:


```
# Interpret args by order
print(divider(7, 22))

# Explicitly specifying values for each arg
print(divider(y = 7, x = 22))
```

## No arguments

It is also possible to write functions that don't take _any_ arguments:

```
def gimme5():
    return 5

gimme5()
```

Although functions taking no arguments may not seem of obvious use, I use them
often.

For example, I often have a `load_data()` function (or some variation on this),
which takes no arguments and returns some data associated with my project.

## Default arguments

We can also construct functions that take _default_ arguments.

Default arguments are used when no value is given for a variable:

```
def weather_forecaster(weather = "rain"):
    return "Tomorrow will be: " + weather
```

If we call this function _without_ passing a value through for `weather`, our
function will use the default value:

```
weather_forecaster()
```

However, we can "override" default arguments by passing a value through:

```
weather_forecaster("grey")
```

### Multiple arguments with defaults

If we wish to define a function that takes _multiple arguments_, some of which
are _optional_, we must specify the default arguments last:

```
# n takes a default argument and so 
# must be specified after x
def to_the_power(x, n=2):
    return x ** n
```

We call this function as usual:

```
# No value for n. Function will use default
to_the_power(4)
```

```
# Value passed through for n, will override default
to_the_power(4, 3)
```

## Good practices

Great, so we can write functions.

But can we write functions that are well-documented, intuitive, and easy to
(re-)use?

I promise, although your function may make perfect sense to your present self, it often
makes a _lot less sense_ to your future self. (Speaking from many experiences of
being sabotaged by my past self.)

Let's see a couple of tips for writing better functions.

### Docstrings

We use _docstrings_ to add documentation to our functions that can then be
accessed from the `help()` function.

Docstrings must be specified at the top of our function's code block, and be
surrounded by triple quotation marks:

```
def to_the_power(x, n=2):
    """This is a docstring."""
    return x ** n

help(to_the_power)
```

Docstrings are useful not just for other people who may read your code
(colleagues, friends, enemies, etc.), but for your future self.

It is good practice to use your docstring to give a quick description of your
function's purpose, and to detail any arguments that your function takes, and
what types they should be.

The docstrings of the functions of the numpy package are a gold-standard
to aim for:

```
import numpy as np
help(np.sum)
```

Let's give our function a numpy-style docstring:

```
def to_the_power(x, n=2):
    """
    Raise x to the power of n

    Parameters
    ----------
    x : float, int, or np.ndarray
        The base value to exponentiate.
    n : float, int, optional
        The exponent to raise the base to. The default is 2.

    Returns
    -------
    float, int or np.ndarray
        x raised to the power of n.
    """
    return x ** n

help(to_the_power)
```

### Aside: another way of accessing help

Although we've previously used the `help()` function to view documentation,
most modern Python IDEs allow a more convenient syntax using `?`:


```
to_the_power?
```

Using `?` formats documentation nicer than `help()` does.

We can also use `??` to view the source code of functions:

```
to_the_power??
```

However, `??` won't reveal a lot for functions that in turn call C code (like
much of the numpy package):

```
np.sum??
```

### Naming

Try to choose function names that are indicative of your function's purpose.

```
# Bad function name
def my_func(x):
    """Compute the arithmetic mean of the list x."""
    return sum(x) / len(x)

my_func([57, 49, 27, 35])
```

```
# Better function name
def compute_mean(x):
    """Compute the arithmetic mean of the list x."""
    return sum(x) / len(x)

compute_mean([57, 49, 27, 35])
```
