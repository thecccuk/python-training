def combine_sectors(files):
    """
    Combine per-sector emissions csv files into a single DataFrame.

    Parameters
    ----------
    files : list of str
        Filenames to combine. Non-csv files are ignored.

    Returns
    -------
    pandas.DataFrame
        The concatenated data, with a `sector` column taken from each
        filename.
    """
    df = pd.DataFrame()
    for file in files:
        if file.endswith(".csv"):
            sector_data = pd.read_csv(file)
            sector_data["sector"] = file.replace(".csv", "")
            df = pd.concat([df, sector_data])
    return df
```

def combine_sectors(files):
