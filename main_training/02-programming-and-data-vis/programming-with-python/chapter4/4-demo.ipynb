# Python Scripts

All of the code we have covered so far has been presented in
Jupyter notebooks. Notebooks have many advantages such as:

- Interactive, cell-by-cell execution: great for exploring data without
  rerunning everything.
- Inline output: plots, tables, and images appear right next to the code
  that made them.
- Good for documenting your thought process: you can mix code, results,
  and markdown in explainable reports.
- Great learning aid (like in these courses!)

But they also have drawbacks when it comes to real world projects:

- Poor version control: diffs in .ipynb files are messy in Git.
- Difficult to reuse functions and code defined in a notebook.
- Notebook cells can be run in any order making notebooks prone to human
  error.
- Difficult to apply code stylers and auto-formatters to notebooks (more
  on this in the Python Best Practices course).

This is why most Python projects rely on **Python scripts**.

## Structure of a Python script

A Python script is really just a plain text file with a .py extension.
All of the Python code that we have seen in these courses (variables,
functions, loops, conditional code, comments) can live in a script.

The basic structure of a Python script (often referred to as a **module**)
is provided below.

```python
"""Module docstring explaining what the script is for"""

# Package imports at the top
import package1
import package2


# Global constants
X = "..."
Y = "..."


# Code split up into functions
def function1(...):
    """
    Docstring
    """
    do_something


def function2(...):
    ...
```

## Importing functions from a script

By splitting code up into callable functions, we can then import these
into a different script or notebook. If the script is called
**"module.py"**, just run:

```python
import module
```

Functions defined in the script can then be called with
`module.function1()` and so on. Alternatively we could import a specific
function:

```python
from module import function1
```

## Running a script

We can run a script from the command line using `python module.py`.
If working in an IDE like VS Code or Positron, we can also click a
"Run" button that would normally be visible above the script.

Note that the example script provided above would not actually *do*
anything if we ran it, aside from importing some packages and
defining some constants and functions.

To actually get an output, we could include some additional lines at
the end of the script.

```python
function1()
function2()
```

The issue here is that these lines will also run when the script is
imported, which can be quite noisy if we're just looking to import
some functions. The better practice is to wrap executable code
inside a special conditional block.

```python
if __name__ == "__main__":
    # Only executes when the script is run
    function1()
    function2()
```

Now, the code only executes when we *run* the script, not when we
*import* it.

## When to use notebooks?

Python scripts have many advantages:

- Clean version control, easy diffs.
- Reusable: easy to import functions elsewhere.
- Straightforward to test and integrate with stylers and formatters.
- Runs the same way every time (no out-of-order execution issues).

But they do also have some drawbacks:

- Slower iteration loop: you often rerun the whole script to see the
  effect of a change.
- No inline visual output: need to save plots to files or use a
  separate viewer.
- Less convenient for quick, ad hoc exploration.

As a rule of thumb: use notebooks for exploration, prototyping, and
reporting; move to scripts once you need something reliable, testable,
and reusable.
