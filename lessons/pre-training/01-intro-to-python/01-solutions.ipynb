Question 4a) Define a list with a set of five numeric values between -5 and 5 (they can be decimals if you want). Call this list co2_changes.

````python
# example solution - the numbers you choose to have in your list are completely up to you!

co2_changes = [0.8, 2.3, 1.3, 0.4, -0.5]
````

Question 4b) Define a function called `update_co2_concentration` which does the following:
- Takes a value for the starting atmospheric concentration of co2. Call it `starting_co2_concentration`.
- Takes a value for an additional amount of co2 added. You can call this argument `additional_co2`.
- Returns the sum of these two inputs

````python
# solution

def update_co2_concentration(starting_co2_concentration, additional_co2):
    # optional print statement to tell you what your function is doing
    print("Updating CO2 concentration...")
    return starting_co2_concentration + additional_co2
````

Question 4c) Write a `for` loop which runs through all of the values in your `co2_changes` list, and updates your `co2_concentration` variable with the values in that list, using the `update_co2_concentration` function you have designed above. Write a print statement in your loop which tells you what the new value of the `co2_concentration` variable is at each iteration.

````python
# solution 

for change in co2_changes:
    co2_concentration = update_co2_concentration(co2_concentration, change)
    print(f"New CO2 concentration is {co2_concentration:.1f}")
````

Bonus: not what was asked for but a better approach to the question above would probably be to keep the updated co2 concentrations in a separate variable so we can refer back to the initial value if we need it. An approach which does this is set out below.

*Note: The ```:1.f``` that you see in the answer is a way of specifying the number of decimal places to show for the number preceeding it.*

````python
initial_co2_concentration = co2_concentration

for change in co2_changes:
    co2_concentration = update_co2_concentration(co2_concentration, change)
    print(f"New CO2 concentration is {co2_concentration:.1f}")

concentration_change = co2_concentration - initial_co2_concentration

print(f"The total concentration change is {concentration_change:.1f}.")
print(f"If we've done this correctly, this should be equal to the sum of our listed values, which is equal to {sum(co2_changes):.1f}.")
````
