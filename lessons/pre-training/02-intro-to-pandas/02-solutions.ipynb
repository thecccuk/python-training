Question 2c) Calculate a new column which holds the sum of the emissions and removals columns.
- Call this column ```net_emissions_gtco2```

````python
# solution below

projected_df["net_emissions_gtco2"] = projected_df["emissions_gtco2"] + projected_df["removals_gtco2"]
````

Question 2d) Using the conversion factor below, create a new column with the change in co2 concentration resulting from net emissions, measured in ppm.
- The conversion factor is 1 ppm = 7.8 GtCO2
- You should call this column concentration_change_ppm

````python
# solution below

CONVERSION_FACTOR = 7.8

projected_df["concentration_change_ppm"] = projected_df["net_emissions_gtco2"] / 7.8
````

Question 3a) Add a new column to your dataframe to hold the initial co2_concentration value (before applying the change in concentration).
- Use the final, 2025 value from your conc_df dataframe for the whole column.
- Call this column ```initial_conc_ppm```.

````python
# solution below

initial_co2_concentration = conc_df.loc[conc_df["year"] == 2025]["co2_ppm"].values[0]

projected_df["initial_co2_ppm"] = initial_co2_concentration

projected_df
````

Question 3b) Create a cumulative sum column from your concentration change column which we can then use to get final co2 concentrations.
- Use the ```pd.Series.cumsum()``` method.
- Call this column ```concentration_change_cumsum```.

````python
# solution below

projected_df["concentration_change_cumsum"] = projected_df["concentration_change_ppm"].cumsum()

projected_df
````

Question 3c) Create a new column with the final co2 concentrations after accounting for the cumulative changes.
- This means you need to sum the concentration and cumulative change columns.
- Call this column co2_conc.

````python
projected_df["co2_ppm"] = projected_df["initial_co2_ppm"] + projected_df["concentration_change_cumsum"]

projected_df
````

Question 4a) Concatenate your conc_df and projected_df dataframes.
- Store the new dataframe in ```full_timeseries_df```.
- Use pd.concat() to achieve this. Look up the documentation to understand the syntax.

````python
# solution below

full_timeseries_df = pd.concat([conc_df, projected_df])
````

Question 4b) Let's make this just a timeseries of CO2 concentrations. To do that, drop all columns containing NA values.
- Hint: you can use ```pd.DataFrame.dropna()``` to achieve this.
- Look up the syntax of this method online (or ask Copilot).

````python
full_timeseries_df = full_timeseries_df.dropna(how="any", axis=1)

full_timeseries_df
````

Question 4c) Export this data to a new csv file in the "./data/outputs" folder.
- Use ```pd.to_csv()```.
- Call your file "full_co2_timeseries_1959_2060.csv"

````python
# solution below

full_timeseries_df.to_csv("./data/full_co2_timeseries_1959_2060.csv",
                          index=False) # optional but preferred to get rid of ascending index
````
