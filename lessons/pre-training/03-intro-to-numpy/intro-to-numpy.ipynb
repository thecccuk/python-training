## What is numpy?

In a previous session, we covered the python `pandas` library. We expect that much of our modelling work will involve using `pandas`, as it is the most popular data library for manipulation of tabular data.

However, much of the core calculation and data manipulation logic in pandas is actually defined in another library - the `numpy` library. You might want to think of `numpy` as the engine, on top of which `pandas` adds a load of functionality to simplify and enhance data manipulation.

`numpy` (**num**erical **py**thon) is a very powerful library for performing mathematical operations on ordered lists of numerical data, which are referred to as "arrays". The `numpy.ndarray` is the core object which defines the properties and functionality associated with these data structures.

## Why bother with numpy at all?

In many scenarios we do not in fact need to use numpy, as pandas provides an easier interface with our data, and provides functionality that we need. However, the additional functionality does slow things down, which means that in some cases we might prefer to use numpy. Some examples of best uses are set out below:

- `numpy` is better with:
    - clean, purely numerical data
    - vector and matrix data structures
    - maths-heavy operations like linear algebra or optimisation


- `pandas` is better with: 
    - messy, incomplete, and mixed type data
    - tabular data, which comes from sources like CSV, excel, SQL etc
    - finicky data cleaning and manipulation operations

Let's take a whistlestop tour through some numpy functionality.

```python
# first, installing and importing the numpy library
# note that numpy is a core dependency of pandas, so you will already have this installed if
# you've installed pandas
# This DOES NOT mean that you don't have to IMPORT numpy separately if you want to access its functionality
# directly

! pip install numpy

import numpy as np
```

```python
# let's generate some numpy arrays and take a look

x = np.arange(1,101,1)
y = np.linspace(500, 40000, 100)
```

```python
# showing these arrays

print(type(x))
print(type(y))

x
```

A key thing to understand about numpy arrays of this shape is that they are vectors, mathematically, and behave as such.

```python
# numpy arrays support vectorized operations
# this means that performing an operation on a numpy array
# automatically applies it to every single element in that array
# for example:

x * 2
```

```python
# when we perform an operation between two arrays
# we get the element wise operation

x / y
```

```python
# let's do some operations on said data
# some common mathematical operations on the full arrays below

print(np.mean(x))
print(np.max(y))
print(np.sum(x))
```

```python
# and some other operations below

np.sin(x)
```

```python
# and another

np.sqrt(x)
```

```python
# we can pack these together into a 2D array
# this is now a matrix
# but not that the type is still ndarray
# that's what the nd is about!


arr = np.array((x, y))

print(type(arr))
arr
```

```python
# reinforcing that point about dimensions
# let's look at the shape attribute for our x
# and arr variables

print(x.shape)
print(arr.shape)
```

```python
# we can transpose our arr ndarray
# to get it reshaped as a 100 by 2 matrix

arr = arr.T
```

```python
print(arr.shape)
```

```python
# numpy interacts very easily with pandas
# so we can switch between the two as it suits us to

! pip install pandas

import pandas as pd
```

```python
# creating a dataframe from our arr ndarray

df = pd.DataFrame(data=arr, columns=["x", "y"])

df
```

```python
# creating a new column

df["z"] = df["y"] ** df["x"]
```

```python
# turning that into a numpy ndarray

z = df["z"].to_numpy()

z
```
